# Breadboard

```{figure} images/breadboard-horizontal-rows.png
---
name: Breadboard horizontal rows
---
Horizontal rows of contacts
```
Horizontal rows of contacts on a breadboard are connected together, which allows for easy wiring of components.

```{figure} images/breadboard-vertical-power.png
---
name: Breadboard vertical power
---
Vertical power rails
```
On each side of the breadboard there are power rails. These can be connected to the power source and used to power components on the breadboard.

```{figure} images/breadboard_ic.png
---
name: Breadboard with IC
---
IC on a breadboard
```
A breadboard with an integrated circuit (IC) is a very common breadboard use in electronics. Thanks to a gutter in the middle of the breadboard, the IC can be connected and not short circuited.

<small>Images source: [https://computers.tutsplus.com/tutorials/how-to-use-a-breadboard-and-build-a-led-circuit--mac-54746](https://computers.tutsplus.com/tutorials/how-to-use-a-breadboard-and-build-a-led-circuit--mac-54746)</small>

You can make a circuit with on board voltage supplies and measure with the voltmeter. All components are placed on the breadbaord. This can be done without connecting the ALPACA to your laptop. The breadboard has the pins on a row (in the middle) and column (on the edges) connected. 

```{figure} ../../images/alpaca.avif
---
name: ALPACA board
height: 300px
---
ALPACA setup
```
