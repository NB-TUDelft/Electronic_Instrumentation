# Introduction to the oscilloscope (scope)
The oscilloscope is an indispensable device to test electronic circuits and is an important complement to the DMM. 

The DMM provides a reading of a circuit at a single point in time. This means that if you are trying to measure a signal that is changing over time, the DMM won't be able to visualise it. When you measure an alternating current (AC) signal such as a sine wave using the DMM, the reading will only show the $U_{rms}$ of the signal. 

For AC signals, it may be more interesting to see how the voltage changes over time, rather than just the $U_{rms}$. This would give us greater insight into the shape and frequency of the signal. This is exactly what the **oscilloscope (often called scope for short)** allows us to do! The oscilloscope that we will be using during this course is the *Siglent SDS800X HD Series*, as seen in the image below.

## Basics of use

```{tip}
The oscilloscope display is touchscreen.
```
    
### Switching channels on/off
After turning on the oscilloscope and connecting signals via the input channels, you can switch the channels on/off to see or hide the signals on the display. To turn a signal on/off:
* Press the coloured button with the number corresponding to the input channel connected to the signal. 
* Once switched on, you should see the yellow/pink/blue/green coloured signal on the screen and the display settings of that channel in the bottom left corner. 

### Channel settings
To adapt the settings for a specific channel, press the coloured button with the number corresponding to the channel whose settings you want to change. This will cause a menu bar to appear on the right side of the display, where you can alter settings for the following:
* Channel on/off.
* Coupling: when DC coupling mode is selected, the signal is displayed unchanged, when AC coupling mode is selected, the average value of the signal is subtracted before display. 
* Probe: this allows you to compensate for the attenuation factor of the scope probe. See [Scope probe](intros:scope_probe) for more information. 
* Bandwidth limit


### Scaling of voltage and timescale
Adjust the vertical and horizontal scale to show the signal clearly on the display.
* The **Vertical** knob tunes between V $\leftrightarrow$ mV. Thus, turning clockwise zooms in along the voltage axis.
* The **Horizontal** knob tunes between s $\leftrightarrow$ ns. Thus turning clockwise zooms in along the time axis.

### Basic triggering
The trigger settings determine when the oscilloscope starts to display a waveform. The oscilloscope acquires data while waiting for the trigger condition to occur. This trigger condition is the measured voltage crossing a set threshold with a certain slope. After a trigger is detected, the oscilloscope continues to acquire enough data to draw the waveform on the display. The basic trigger settings include:
* Source: here you can select a channel for the oscilloscope to use when triggering.
* Slope: the slope that the input signal must have for the oscilloscope to trigger. This can be set to Rising or Falling. 
* Trigger level: this is the threshold that the input signal must reach for the oscilloscope to trigger. If the trigger level is never reached, then the oscilloscope cannot trigger. This results in the signal jumping around along the time axis.

For example, if a 4 $U_{pp}$ sine wave is connected to channel 1 of the oscilloscope and you select the trigger source as channel 1, slope as Rising and the trigger level as 1 V, you would expect to see the following signal.

```{figure} images/Trigger_rising_slope.png
---
width: 400px
name: Trigger rising slope
---
A 4 Upp sine wave viewed with a 1 V trigger level and rising slope on the oscilloscope
```


To configure the trigger settings on the oscilloscope:
* Press **Setup** on the trigger settings region of the oscilloscope. Alternatively, you can press **Trigger** on the menu bar at the top of the oscilloscope display. This should cause a box to appear on the right of the oscilloscope display with the various trigger settings.
* From here you can select the trigger source and slope.
* Turn the **Level** knob found on the trigger settings region of the oscilloscope to adjust the trigger level or push the **Level** knob to set the trigger level to the average value of the signal you selected as the trigger source. 

## Cursors
Cursors allow you to measure points along the time and/or voltage axes and the distance between these points. To configure cursors:
* Push **cursors** to switch on cursors. This will cause the cursors setup menu to open on the display. From here you can select:
  * Source (any of the input channels or Math).
  * Type: x (time) or y (voltage).
* Turn the **Intensity/Adjust** knob to move the selected cursor. 
* Push the **Intensity/Adjust** knob to select a different cursor. 

## Measure
Instead of the manual cursors, you can also use the Measure functionality of the oscilloscope to automatically determine key features of the signal, such as the $U_{pp}$ or the frequency. 
* Press **Measure**.
* Default values, such as Pk-Pk, Mean, Period and Freq are measured and will be displayed at the bottom of the oscilloscope display. 
* Select additional measurement parameters under Type. 

## Fourier transform
The Fourier transform allows you to view the component frequencies of a signal. To observe the Fourier transform (FFT) on the oscilloscope:
* Press **Math**. This will cause a menu bar to appear on the right side of the oscilloscope Display. 
* Select FFT under Function.
* Select the channel whose Fourier transform you would like to view under Trace. 
* Via Math > Horizontal or Vertica, the axis of the FFT plot can be adapted. The original V-t diagram can still be apated by the Vertical and Horizontal turning knobs. 

On the scope you only see the positive frequency axis. Hence, for a sine wave, you would only expect to see one peak on the FFT plot. 

```{tip}
Try to visualise the FFT of a block wave on the scope.
```

## Scope probe
The scope probe is a type of cable used to measure signals on the oscilloscope. While the scope probe and coax cable share many similarities, the coax cable cable has a slight low-pass filtering effect. This represents a disadvantage when measuring high frequency signals, as the coax cable will attenuate (reduce the intensity) of the signal it is measuring. The scope probe mitigates this issue, as it contains an additional capacitor to offset this low pass filtering effect. As this low pass filtering effect is quite slight, if the circuit you are measuring is already a low pass filter, then there will be no noticeable difference between using a coax cable and the scope probe, because the low-pass filtering effect of your circuit will likely be much greater than the effect of the coax cable. However, in cases where the bandwidth of your measurement is limited by using a coax cable, it is beneficial to opt for using the scope probe.

### Probe compensation

A 10X scope probe has a built-in **trimmer (adjustable) capacitor**, $C_{comp}$ in its tip. Together with the probe's 9 MΩ series resistance and the scope's input impedance (1 MΩ ∥ $C_{in}$), it forms a compensated voltage divider. For a flat frequency response, the time constants must match:

$$R_{probe} \cdot C_{comp} = R_{in} \cdot C_{in}$$

The value of this compensation capacitor must be set to offset any filtering effects from the scope probe. To correctly adjust the value of the compensation capacitor:
* Connect the hook of the scope probe to the oscilloscope's **CAL output** and the crocodile clip to the ground terminal, as seen in the image below. This should cause a block wave to appear on the oscilloscope display.
* Set the channel to 10X and ensure that the red slider on the scope probe is also set to 10X.
* Insert the screwdriver tool found with the scope probe and adjust until the square has flat tops.
