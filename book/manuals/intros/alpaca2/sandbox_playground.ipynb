# Sandbox and Playground

Sandbox and playground areas on your ALPACA provides different tools and functionalities for your own custom circuits.

## Playground

### Switches

The switches share common rails and resistors for pull-up/pull-down configurations, where the common rail lets you connect all switches to either ground or a voltage source.

This way, in both pull-up and pull-down configurations, you only need separate cables for each signal output.


Relay is essentially a switch that can be controlled digitially via the `RELAY` pin. 

**When `RELAY` Pin is unpowered:**

`COM` pins connect to their respective `NC` (normally connected) and decouple from `NO` (normally open)

**When `RELAY` Pin is powered:**

`COM` pins connect to their respective `NO` and decouple from `NC`

### Variable U

Uvars presented at the bottom right of the board are simply knobs that can be turned to output a voltage between 0-3.3V. They output their voltage at their `Uvar` pin found below.

### BNC Connectors

BNC connectors allow for coax cables to be connected to board to couple up with instruments like the oscilloscope.

In BNC connections, the outer conductor **(the shield) is ground**. It carries the cable's braided shield and connects to the signal return/ground. **The center pin carries the signal.**

Therefore, in most cases you should connect your input/output signal to the `Signal` pin and `Shield` should be connected to `GNDA` via a jumper.

#### Connecting to oscilloscope

You might want to inspect you signal in a timely manner instead of just the voltmeter. The oscilloscope exists for this reason!

There are two manners possible:

<font color="#ff0066">**Option 1: Coax cable**</font>

You can plug a Coax cable between BNC connectors of oscilloscope and ALPACA. As previosuly mentioned whatever is connected to your `Signal` pin will appear on the oscilloscope.

<font color="#ff0066">**Option 2: Scope Probe**</font>

The scope probe's ground should be connected to the plate that says *Oscilloscope Ground* found at the bottom of ALPACA. You can use the probe's hook to inspect signals. You can read more about the probe in the [Oscilloscope chapter](../oscilloscope.ipynb/#scope-probe).
