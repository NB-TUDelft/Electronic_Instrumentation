## Live Plotting

:::{admonition} Background: 'yield'
:class: note, dropdown

The `yield` keyword is a new concept. It is simillar to the `return` keyword but it doesn't stop the function. Instead it **yields (adds) the value to the output and continues**. This can be compared to lists.

```python
def foo():
    l = []
    for i in range(5):
        l.append(i + 1)
    
    return l

output = foo()
```

gives the exact same result as

```python
def foo():
    for i in range(5):
        yield i + 1

# foo() returns a 'generator' which is essentially
# something that can be used to generate a list
output = list(foo())
```

Where they start to differ is when you start to use them in loops:

```python
def foo():
    for i in range(5):
        print("Here ", i)
        yield i + 1

for j in foo():
    print("There ", j)

### OUTPUT
# Here 1
# There 2
# Here 2
# There 3
# ...
```

```python
def foo():
    l = []
    for i in range(5):
        print("Here ", i)
        l.append(i + 1)
    
    return l

for j in foo():
    print("There ", j)

### OUTPUT
# Here 1
# Here 2
# Here ..
# There 1
# There 2
# There 3
# ...
```

As you can see the main difference is with `yield` the function doesn't stop it continues. Where as returning needs the whole function the be finished.
:::


When acquiring data, standard plots fall short as they show you the results at the very end. With **live plotting**, you can update an active graph. From the below example you can see how to "animate" matplotlib graphs to match your output.

```python
import matplotlib.pyplot as plt

@pico.task
def measure_samples(N, dur):
    '''
    Take measurements for dur seconds and N samples for each measurement
    '''
    from utime import ticks_add, ticks

    deadline = ticks_add(ticks(), dur) # dur in the future

    while ticks_diff(deadline, ticks()) =< 0:
        total = 0
        for i in range(N):
            total += adc0.read_u16()
        
        yield total * 3 / 2**16 / N # Add average to the list

    return # Stop

plt.ion()   # Enable matplotlib
fig, ax = plt.subplots() # Create figure and an axis

t = []  # Time
y = []  # Values
# Initialize an empty line object
(line,) = ax.plot([], [])

# This for loop will run as long as 'yield' is 'ran'
for val in measure_samples(50, 5):
  t.append(time.time()) # Add current time
  y.append(val) # Add measurement

  ax.relim() # Adjust the x-axis and y-axis
  ax.autoscale_view()

  # Update the line with current data
  line.set_ydata(y)
  line.set_xdata(t)

plt.ioff()
plt.show()
```
