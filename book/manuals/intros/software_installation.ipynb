# Software Installation

This course requires the following software on your computer:

- Conda
- VS Code

If you do not have these software installed on your machine, you can follow section [2.1 Software Installation](https://interactivetextbooks.tudelft.nl/programming-foundations-2ed/content/chapter2/installation.html) of the NB1420 Programming Foundations interactive textbook.


### Creating a New Conda Environment

For this course, you will need a separate conda environment called `ei-env`:

<span style="background:#ff1493;color:#fff;padding:2px 8px;border-radius:4px;font-weight:700;font-size:0.85em;box-shadow:0 0 6px rgba(255,20,147,0.6);">🚩 TODO: create ei-env.yaml and link it</span>


1. **Download the `ei-env.yaml` file and save it.**

2. **Go to the directory:** Open Terminal (macOS) or Command Prompt (Windows), then change directory to the location of `ei-env.yaml`.


For example, if it is located in your Downloads folder, run:

```sh
cd ~/Downloads
```

3. **Create the environment using the following command:**

```sh
conda env create -f ei-env.yaml
```

If the following message appears then you have successfully completed this step:

```sh
#
# To activate this environment, use
#
#      $ conda activate ei-env
#
# To deactivate an active environment, use
#
#      $ conda deactivate
```

The required packages are installed on your machine, and you are ready to use the ALPACA!

```{admonition} Changing environment in VS Code
:class: tip

If VS Code is complaning about modules `belay`, `machine` or etc. not being found then you have to switch the Python environment by pressing `Ctrl/Cmd+Shift+P` and typing in `Python: Select Intepreter`. In the dropdown select `ei-env`.
```
