#### Task B3: Digital input and output pins
The ALPACA's student Pi-Pico contains both digital input and digital output pins. These pins are referred to as *digital*, as they are only able to be in one of two states, either high (3.3 V) or low (0 V). This is in contrast to analogue pins, which can read or output a continuous range of voltages. 

Digital output pins output 3.3 V when set to True and output 0 V when set to False. Similarly, a digital input pin can determine whether 3.3 V or 0 V are connected. When 3.3 V are connected to a digital input pin, it will be assigned as True and when 0 V are connected, it will be assigned as False. 

To programme a digital input or output pin, you first need to import the `Pin` function from the `machine` MicroPython module. This allows you to define pins. You then need to define a pin using the pin ID and state whether the pin is an output or input pin. 

```{tip}
Pin IDs digital output pins: 10, 11, 12, 13

Pin IDs digital input pins: 14, 15, 16, 17 

You can also find the pin IDs in the ALPACA manual.
```

Once a digital output pin is defined, it can be programmed as True to output 3.3 V, or as false to output 0 V and once a digital input pin is defined, it is possible to check whether it is connected to 3.3 V or 0 V. 


For example, pin 10 on the Pi-Pico is a digital output pin and pin 14 is a digital input pin. In the code cell below, you can see how these pins can be defined and programmed. 

```{note}
The names Dout1 and Din1 were selected, as they are easy to remember. You can assign different names to the pins. 
```

````python
from machine import Pin #Import the module used to define pins

Dout1 = Pin(10, Pin.OUT) #Give the pin a name. Define the pin using the pin ID (in this case, the ID is 10). State whether it is an input/output pin
Din1 = Pin(14, Pin.IN) #Give the pin a name. Define the pin using the pin ID (in this case, the ID is 14). State whether it is an input/output pin

Dout1.value(True) #The value of pin 10 is assigned as True, so it will output 3.3 V
Dout1.value(False) #The value of pin 10 is assigned as False, so it will output 0 V

isPressed = Din1.value() #The variable isPressed will be True when 3.3 V are connected to pin 10 and False when 0 V are connected to pin 14
````

#### Task B4: Switches and pull-up/pull-down resistors
Analogous to the mallet which is used to hit the moles in the Whack-a-mole game, you will use switches to turn off the LEDs. A switch is a valuable electronic component which can be used to close or open a path for current flow. A button is a special type of switch that works when pushed. 

As seen in the figure below, the ALPACA has four buttons.

```{figure} images/switches_ALPACA.png
---
width: 400px
name: Schematic of switches on the ALPACA
---
Schematic of switches on the ALPACA
```
As you can see in this figure, the circuit containing the button also contains a $10 kΩ$ resistor. 

We have two different options for how we can connect the button with respect to the resistor. 

In the figure shown below, the resistor is connected in such a way that we refer to it as a **pull-up resistor**.

```{figure} images/Pull_up_resistor.svg
---
height: 200px
name: Pull up resistor
---
Pull-up resistor configuration
```
In this configuration, when the button is not pressed, the digital input pin will be connected to 3.3 V. This means that the value of the pin will be assigned as True by default. When the button is pressed, the digital input pin will be connected to ground (0 V). The $10 kΩ$ resistor is connected between the 3.3 V supply and ground to limit the flow of current, preventing a short circuit when the button is pressed.

You can also connect the resistor as a **pull-down resistor**, as seen in the figure below.

```{figure} images/Pull_down_resistor.svg
---
height: 200px
name: Pull down resistor
---
Pull-down resistor configuration
```
In this configuration, when the button is not pressed, the digital input pin will be connected to ground (0 V). This means that the value of the pin will be assigned as False by default. When the button is pressed, the digital input pin will be connected to 3.3 V. The $10 kΩ$ resistor is connected between the 3.3 V supply and ground to limit current flow when the button is pressed, preventing a short circuit. 

```{note}
In summary:
- Pull-up resistor configuration: Default value of the digital input pin (when the button is not pressed) is 3.3 V or True. 
- Pull-down resistor configuration: Default value of the digital input pin (when the button is not pressed) is 0 V or False.
- In both configurations, the $10 kΩ$ resistor limits the flow of current when the button is pressed, preventing a short circuit.
```
