#### Task B1: Decide whether a diode conducts
```{admonition} Estimated time: 15 min
:class: estimated-time no-content
```

A diode has an **anode** and a **cathode**. Conventional current flows from anode to cathode when the diode is forward biased, and the diode blocks current in the other direction. Before you can solve a diode circuit you have to choose a model for that behaviour.

* **Ideal model:** a conducting diode has zero voltage across it. A blocking diode is an open circuit.
* **Constant-drop model:** a conducting silicon diode has $U_A - U_C \approx 0.7\,\mathrm{V}$. A blocking diode is an open circuit.
* **Real model:** the current rises roughly exponentially with the forward voltage, so the forward voltage depends on the current and on the temperature. LTspice uses this model through a component model such as the 1N4148.

```{attention}
A conducting diode is not a wire when you use the constant-drop model. Keep the $0.7\,\mathrm{V}$ in every voltage equation.
```

Use these two questions whenever you meet a diode:

1. What are $U_A$ and $U_C$ if the diode blocks?
2. Does $U_A - U_C$ forward bias the diode under your chosen model?

If the answer to the second question is yes, replace the diode by its conducting model and solve the circuit again.

##### Worked example: the shunt clipper

{numref}`fig-shunt-clipper` is the first circuit of this manual. The source drives a series resistor $R_1$, and the diode sits from the output node to ground with its **anode at the output**. Use $R_1=1\,\mathrm{k}\Omega$ and

$$
U_{in}(t)=1\,\mathrm{V}\sin(2\pi\cdot 1\,\mathrm{kHz}\cdot t).
$$

$R_L$ is the oscilloscope input, so $R_L\gg R_1$ and you may treat it as an open circuit.

```{figure} images/shunt-clipper.svg
---
name: fig-shunt-clipper
height: 310px
---
The shunt clipper. The diode anode is at the output node and the cathode is at ground.
```

Start with the **ideal model** and answer the two questions.

1. Assume the diode blocks. No current flows through $R_1$, so $U_{out}=U_{in}$. The anode is at $U_A = U_{in}$ and the cathode at $U_C = 0\,\mathrm{V}$.
2. An ideal diode is forward biased as soon as $U_A - U_C > 0$, so as soon as $U_{in} > 0\,\mathrm{V}$.

So the assumption only holds while $U_{in}\leq 0\,\mathrm{V}$. In that half of the cycle the output follows the input. In the other half the diode conducts and, being ideal, holds the output at $0\,\mathrm{V}$ while the whole input appears across $R_1$:

$$
U_{out}\approx
\begin{cases}
U_{in}, & U_{in}\leq 0\,\mathrm{V},\\
0\,\mathrm{V}, & U_{in}> 0\,\mathrm{V},
\end{cases}
\qquad
I_D=\frac{U_{in}-U_{out}}{R_1}.
$$

At the positive peak the ideal model predicts $I_D = 1\,\mathrm{V}/1\,\mathrm{k}\Omega = 1\,\mathrm{mA}$.

That is the whole method: assume a state, check the bias, then solve. In your notepad, sketch this ideal prediction over one full period, with the zero line and both voltage axes marked. In [Task A1](Task_A1_5_1) you repeat the two questions with the constant-drop model, which moves the clipping level away from zero.

````

(Task_A4_5_1)=
#### Task A4: Choose a smoothing capacitor
```{admonition} Estimated time: 10 min
:class: estimated-time no-content
```

A capacitor across $R_L$ charges near each rectified peak and supplies the load between peaks. The result is a mean output with ripple, as shown in {numref}`fig-bridge-ripple`.

```{figure} images/bridge-ripple.svg
---
name: fig-bridge-ripple
height: 300px
---
Peak-to-peak ripple is the voltage the load loses between two capacitor recharge events.
```

Use the **silicon** bridge from [Task A3](Task_A3_5_1): $f=100\,\mathrm{Hz}$, $R_L=1\,\mathrm{k}\Omega$, and a predicted rectified peak of $8.6\,\mathrm{V}$. The design requirement is $U_{ripple,pp}\leq0.5\,\mathrm{V}$.

For a full-wave rectifier, estimate

$$
U_{ripple,pp}\approx\frac{I_L}{2fC}.
$$

1. Use $I_L\approx8.6\,\mathrm{V}/R_L$ to calculate the minimum capacitance.
2. Choose the smallest value that meets the requirement from $1\,\mathrm{\mu F}$, $10\,\mathrm{\mu F}$, $47\,\mathrm{\mu F}$, $100\,\mathrm{\mu F}$, and $220\,\mathrm{\mu F}$.
3. Predict what happens to the ripple if $R_L$ is halved.
4. Predict what happens to the ripple if the source frequency is halved.
5. State two assumptions behind the ripple estimate.

```{admonition} Check your answer
:class: answer dropdown
The estimated load current is $8.6\,\mathrm{mA}$, so

$$
C_{min}\approx\frac{8.6\,\mathrm{mA}}{2(100\,\mathrm{Hz})(0.5\,\mathrm{V})}=86\,\mathrm{\mu F}.
$$

Choose $100\,\mathrm{\mu F}$, which predicts $0.43\,\mathrm{V}$ of ripple. The next value down, $47\,\mathrm{\mu F}$, predicts $0.91\,\mathrm{V}$ and fails. Halving $R_L$ roughly doubles the load current and therefore the ripple. Halving the frequency also doubles the ripple.

The estimate assumes that the load current is nearly constant between two short recharge pulses, that the ripple is small compared with the output voltage, and that the capacitor is recharged all the way to the peak each time. The last assumption needs a source and diode resistance that are small compared with $R_L$, which is exactly what fails in the LED bridge.
```

### Simulate

Use the built-in 1N4148 model for every silicon diode. A real diode does not switch at one exact voltage, so to compare circuits consistently, define the simulated transition as the point where $|I_D|$ first exceeds $0.1\,\mathrm{mA}$.

```{attention}
A bridge rectifier has separate source and load node pairs. In LTspice, ground one source terminal and leave both load terminals floating. Plot the differential load voltage as $V(out\_plus,out\_minus)$. Do not connect a second ground to the negative load terminal, because that shorts one bridge diode.
```

(Task_S1_5_1)=
#### Task S1: Simulate the two clippers
```{admonition} Estimated time: 20 min
:class: estimated-time no-content
```

Build both clippers in one LTspice file:

1. the shunt clipper from [Task A1](Task_A1_5_1), with $R_1=1\,\mathrm{k}\Omega$, $R_L=1\,\mathrm{M}\Omega$, and `SINE(0 1 1k)`;
2. the biased circuit from [Task A2](Task_A2_5_1), with $R_1=R_2=1\,\mathrm{k}\Omega$, a $5\,\mathrm{V}$ supply, and `SINE(0 8 1k)`.

Run a transient simulation over at least five periods with a maximum time step of $5\,\mathrm{\mu s}$.

For each circuit:

1. Plot $U_{in}$, $U_{out}$, and $I_D$.
2. Measure the minimum and the maximum $U_{out}$.
3. Measure the input voltage at which $|I_D|$ first exceeds $0.1\,\mathrm{mA}$.
4. Measure the peak diode current and the diode voltage at that current.
5. Explain one difference between the 1N4148 result and your ideal or constant-drop prediction.

Then change the shunt clipper source amplitude to $5\,\mathrm{V}$ and rerun. Record the new clipping level and the new peak diode current, and check them against your prediction in step 5 of [Task A1](Task_A1_5_1).

Complete this table in your notepad:

| circuit | ideal prediction | constant-drop prediction | 1N4148 simulation | assumption that explains the difference |
|---|---|---|---|---|
| shunt clipper, $1\,\mathrm{V}$ |  |  |  |  |
| shunt clipper, $5\,\mathrm{V}$ |  |  |  |  |
| biased clipper |  |  |  |  |

Save one clearly labelled plot per circuit. You will compare these plots with oscilloscope traces during the practicum.

(Task_S2_5_1)=
#### Task S2: Simulate the silicon bridge and the capacitor design
```{admonition} Estimated time: 15 min
:class: estimated-time no-content
```

Build the bridge from [Task A3](Task_A3_5_1) with four 1N4148 diodes, `SINE(0 10 100)`, and $R_L=1\,\mathrm{k}\Omega$.

1. Simulate without a capacitor for at least $100\,\mathrm{ms}$.
2. Plot the source voltage and $V(out\_plus,out\_minus)$.
3. Measure the rectified output frequency and peak voltage, and compare them with your prediction from [Task A3](Task_A3_5_1).
4. Plot all four diode currents and confirm that the diagonal pairs alternate.
5. Add the capacitor you selected in [Task A4](Task_A4_5_1) across $R_L$.
6. Simulate for at least $300\,\mathrm{ms}$, then measure the mean output and the peak-to-peak ripple over the final two cycles.
7. Decide whether the result meets the $0.5\,\mathrm{V}$ requirement.

If the design misses the requirement, revise one assumption or one component value and run again. Record both results and explain the change. Save the schematic and the plots for the practicum.

(Task_S3_5_1)=
#### Task S3: Simulate the ALPACA LED bridge
```{admonition} Estimated time: 10 min
:class: estimated-time no-content
```

Copy the bridge from [Task S2](Task_S2_5_1) and turn each arm into one ALPACA LED branch: a $390\,\Omega$ resistor in series with an LED. Keep `SINE(0 10 100)` and $R_L=1\,\mathrm{k}\Omega$, and remove the smoothing capacitor.

An LED is not in the standard LTspice diode library, so you supply the constant-drop model yourself. Add these lines as a SPICE directive and set the value of each diode symbol to the matching model name:

```text
.model LED_GRN D(Ron=1 Roff=1G Vfwd=2.2 epsilon=0.1)
.model LED_YEL D(Ron=1 Roff=1G Vfwd=2.0 epsilon=0.1)
.model LED_RED D(Ron=1 Roff=1G Vfwd=1.9 epsilon=0.1)
.model LED_BLU D(Ron=1 Roff=1G Vfwd=3.0 epsilon=0.1)
```

`Vfwd` is the forward threshold, `Ron` the on-resistance, and `epsilon` a small rounding of the knee that helps the solver converge. This is the constant-drop model of [Task B1](Task_B1_5_1), written out for the simulator. If your LTspice version rejects `Vfwd`, use an ordinary diode in series with a DC voltage source equal to $U_f$ instead.

1. Use the colour assignment you chose in [Task A3](Task_A3_5_1).
2. Plot $V(out\_plus,out\_minus)$ and the current in each of the four branches.
3. Measure the peak output voltage of each half-cycle separately. Compare both with your hand calculation.
4. Measure the conduction interval of one branch as a fraction of the half-period, and explain why it is shorter than for the silicon bridge.
5. Now add the $100\,\mathrm{\mu F}$ capacitor across $R_L$ and simulate again. Report the mean output voltage and the ripple, and explain why the mean sits so far below the peak you measured in step 3.

Save the two output plots, with and without the capacitor. You will compare them with the bench result in [Task I4](Task_I4_5_1).

## Practicum

### Implement and investigate

Work in pairs. Before the practicum you predicted and simulated the shunt clipper, the biased clipper, and the bridge rectifier in both its silicon and its LED form. Now you build each of them, drive them with the **bench function generator**, and reconcile prediction, simulation, and measurement.

Divide the work concretely. For each task, one student builds the circuit on the ALPACA breadboard while the other configures the function generator and the oscilloscope. Swap roles before the next task. During the bridge tasks, one student traces and checks each node against the schematic while the other wires the next branch. Never energize a circuit until both of you agree on what is on the breadboard.

```{attention}
Connect **one** ground reference only, at the function generator negative terminal. Do not add a second ground wire from the scope to the negative load terminal of a bridge, because that short-circuits one bridge element and distorts the measurement. Measure a floating bridge output differentially, with one channel on each load terminal and the scope math subtract function.
```

(Task_I1_5_1)=
#### Task I1: Build and reconcile the shunt clipper
```{admonition} Estimated time: 25 min
:class: estimated-time no-content
```

Build the shunt clipper of {numref}`fig-shunt-clipper` on the ALPACA breadboard with a 1N4148 and $R_1=1\,\mathrm{k}\Omega$. The scope input is the load. Drive the circuit from the **bench function generator**, not from the ALPACA DAC, with a $1\,\mathrm{kHz}$ sine at $1\,\mathrm{V}$ amplitude.

1. Work from the schematic, not from a breadboard layout. Identify the anode and the cathode of the diode before you insert it, and check the orientation with the DMM diode-test mode or with the ALPACA component tester.
2. Put channel 1 on $U_{in}$ and channel 2 on $U_{out}$, both DC coupled.
3. Measure the clipping level, that is the flat value of $U_{out}$ at the positive peak.
4. Measure the input voltage at which $U_{out}$ departs from $U_{in}$.
5. Raise the amplitude to $5\,\mathrm{V}$ and measure the clipping level again.
6. Save one labelled screenshot per amplitude, showing $U_{in}$ and $U_{out}$ together.

The clipping level moved when you changed the amplitude. Explain that shift with the diode current, and state which of the three models of [Task B1](Task_B1_5_1) is needed to predict it. If your measurement disagrees with [Task S1](Task_S1_5_1), change one model assumption at a time and note which change closes the gap.

(Task_I2_5_1)=
#### Task I2: Build and reconcile the biased clipper
```{admonition} Estimated time: 25 min
:class: estimated-time no-content
```

Build the biased circuit of {numref}`fig-biased-diode-circuit` with a 1N4148, $R_1=R_2=1\,\mathrm{k}\Omega$, and the $5\,\mathrm{V}$ rail of the ALPACA Energy Hub as the supply. Remember to switch the Energy Hub on. Drive the input from the function generator with a $1\,\mathrm{kHz}$ sine at $8\,\mathrm{V}$ amplitude.

1. Before you connect the generator, measure $U_{out}$ with the DMM. It should read the Thevenin voltage you calculated in [Task A2](Task_A2_5_1). If it does not, find the wiring error now rather than after the scope is connected.
2. Put channel 1 on $U_{in}$ and channel 2 on $U_{out}$, both DC coupled.
3. Measure the input voltage at which $U_{out}$ leaves its blocking-state value.
4. Measure the peak $U_{out}$, and from it the diode drop at the peak current.
5. Save one labelled screenshot.

Compare the diode drop you measured here with the clipping level you measured in [Task I1](Task_I1_5_1) at a $1\,\mathrm{V}$ amplitude. The same part gave two different drops. Record both currents next to both drops. This pair of numbers is the evidence for [Task C1](Task_C1_5_1), so write it down carefully.

(Task_I3_5_1)=
#### Task I3: Build the silicon bridge and meet the ripple requirement
```{admonition} Estimated time: 40 min
:class: estimated-time no-content
```

Build the bridge of {numref}`fig-bridge-rectifier` from four 1N4148 diodes and $R_L=1\,\mathrm{k}\Omega$ on the breadboard. Leave the capacitor out for now.

Before you energize the circuit:

1. Check each of the four diodes with the DMM diode-test mode, and annotate your schematic with the anode and cathode you measured.
2. Confirm that the two source nodes and the two load nodes are four distinct nodes, and that no extra ground wire bridges them.

Drive the bridge from the function generator with a $100\,\mathrm{Hz}$ sine at $10\,\mathrm{V}$ amplitude, and measure $U_{out}=U(out\_plus)-U(out\_minus)$ differentially.

3. Measure the rectified output frequency. Compare it with the source frequency and with your prediction from [Task A3](Task_A3_5_1).
4. Measure the rectified peak voltage and compare it with the two-drop prediction.
5. Swap the two generator leads at the source nodes. Confirm that the load polarity does not reverse. That is the defining property of a full bridge.

Now add the capacitor you selected in [Task A4](Task_A4_5_1) across $R_L$.

```{attention}
If you use an electrolytic capacitor, observe its polarity. The positive terminal goes to `out_plus`. A reversed electrolytic capacitor can fail and vent electrolyte.
```

6. Measure the mean output voltage and the peak-to-peak ripple with the scope measurement functions.
7. Decide whether the result meets the $0.5\,\mathrm{V}$ requirement.
8. Repeat with one capacitor value lower, or with $R_L$ halved, and record the new ripple.

In your notepad, state whether the original design met the specification. If it did not, revise the value, name the assumption that caused the miss, and record the corrected result.

(Task_I4_5_1)=
#### Task I4: Make the bridge current paths visible with the ALPACA LEDs
```{admonition} Estimated time: 30 min
:class: estimated-time no-content
```

Now build the same bridge from the four Playground LED branches of your ALPACA, as in {numref}`fig-alpaca-led-bridge`, using the header pins listed in [Task A3](Task_A3_5_1). Keep $R_L=1\,\mathrm{k}\Omega$ on the breadboard.

```{attention}
Every LED branch already contains its own internal $390\,\Omega$ resistor, so you must not add one, and you must not leave it out of your calculations either. Both terminals of every branch are floating: neither is tied to the ALPACA ground. Do not connect `out_minus` to ground.
```

**Step 1: measure the forward voltage of each branch.** A diode-test reading is not usable here, because the internal $390\,\Omega$ and a forward voltage above $2\,\mathrm{V}$ push most meters over range. Instead, drive one branch from the $5\,\mathrm{V}$ rail through an external $R_s=470\,\Omega$ and measure your way to $U_f$:

1. Measure $U_{Rs}$ across the external resistor and calculate the branch current $I=U_{Rs}/R_s$.
2. Measure $U_{branch}$ across the whole branch, from the anode-side pin to the cathode pin.
3. Calculate $U_f=U_{branch}-(390\,\Omega)I$.

Record $U_f$ **together with the current it was measured at**, for all four colours. Compare the values with the typical numbers you used in [Task A3](Task_A3_5_1).

**Step 2: watch the two paths alternate.** Wire the bridge with the colour assignment you chose in [Task A3](Task_A3_5_1). Drive it from the function generator at $10\,\mathrm{V}$ amplitude and about $2\,\mathrm{Hz}$, slow enough to follow with your eyes.

4. State your prediction out loud before you switch the generator on: which two LEDs light together, and which two light on the other half-cycle?
5. Record which pair lights in each half-cycle and confirm that the diagonal pairs alternate.
6. Reverse the two generator leads and confirm that the same pairs still alternate.

**Step 3: measure the rectified output.** Set the generator back to $100\,\mathrm{Hz}$ at $10\,\mathrm{V}$ amplitude and measure $U_{out}$ differentially.

7. Measure the peak output voltage of **each** half-cycle. They differ. Explain the difference from your $U_f$ measurements.
8. Compare the peak output with the silicon bridge result of [Task I3](Task_I3_5_1) at the same source setting. Account for the missing volts, term by term: two forward drops and two internal $390\,\Omega$ resistors.
9. Add the $100\,\mathrm{\mu F}$ capacitor across $R_L$. Measure the mean output and the ripple, and explain why the mean sits well below the peak from step 7. Compare with [Task S3](Task_S3_5_1).

Your final schematic must show all four internal $390\,\Omega$ resistors and the header pin number of every connection.

### Compare and Conclude

Work in groups of four at your table. Pool your predictions, LTspice plots, and bench measurements, then fill the tables below in your notepad. The completed tables are the core of your onepager report. The goal is not the formulas, which you already wrote down, but the side-by-side view of prediction, simulation, and measurement.

(Task_C1_5_1)=
#### Task C1: Prediction, simulation, measurement matrix
```{admonition} Consider Tasks [A1](Task_A1_5_1), [A2](Task_A2_5_1), [A3](Task_A3_5_1), [S1](Task_S1_5_1), [S2](Task_S2_5_1), [S3](Task_S3_5_1), [I1](Task_I1_5_1), [I2](Task_I2_5_1), [I3](Task_I3_5_1), and [I4](Task_I4_5_1).
:class: note no-content
```

Fill one row per circuit. For each row, write the rule that decides whether a diode conducts, the key predicted feature, the simulated feature, the measured feature, and the single best explanation of any mismatch.

| circuit | conduction rule | predicted feature | simulated feature | measured feature | best explanation of mismatch |
|---|---|---|---|---|---|
| shunt clipper, $1\,\mathrm{V}$ |  |  |  |  |  |
| shunt clipper, $5\,\mathrm{V}$ |  |  |  |  |  |
| biased clipper |  |  |  |  |  |
| silicon bridge |  |  |  |  |  |
| ALPACA LED bridge |  |  |  |  |  |

Then agree on one answer to each of these two questions, and write the answer your table settles on.

1. Across all five rows, at which currents did the $0.7\,\mathrm{V}$ model predict your measurement to within $50\,\mathrm{mV}$, and where did it fail?
2. The LED bridge and the silicon bridge share a topology but not a waveform. Name the two separate device properties that account for the difference, and say which of the two costs you more output voltage.

(Task_C2_5_1)=
#### Task C2: TA check-off
```{admonition} Consider Tasks [A3](Task_A3_5_1), [A4](Task_A4_5_1), [I3](Task_I3_5_1), and [I4](Task_I4_5_1).
:class: note no-content
```

Before you leave, explain the following to your TA without looking at the finished wiring:

1. Trace one bridge current path from the source, through two elements, to the load, and back. State which two conduct and why the load polarity stays fixed.
2. Show the ripple design evidence: the capacitor you chose in [Task A4](Task_A4_5_1), the ripple you measured in [Task I3](Task_I3_5_1), and whether the design met the $0.5\,\mathrm{V}$ specification.
3. State one condition under which the constant $0.7\,\mathrm{V}$ model is adequate and one under which it is not, using your own numbers from [Task I1](Task_I1_5_1) and [Task I2](Task_I2_5_1).

````{admonition} Deep dive: half-wave against full-wave
:class: deep-dive dropdown

Remove three of the four bridge elements and you get the oldest rectifier of all, a single diode in series with the load, shown in {numref}`fig-half-wave-rectifier`.

```{figure} images/half-wave-rectifier.svg
---
name: fig-half-wave-rectifier
height: 300px
---
A series diode passes one part of the input cycle to the load.
```

Predict first, then check:

* For $R_L=1\,\mathrm{k}\Omega$, a 1N4148, and a $100\,\mathrm{Hz}$ sine at $10\,\mathrm{V}$ amplitude, what are the output frequency and the peak output voltage? How do both compare with the bridge of [Task I3](Task_I3_5_1)?
* The capacitor now has a full period, not a half period, to hold the load up. Rework the estimate of [Task A4](Task_A4_5_1) with $U_{ripple,pp}\approx I_L/(fC)$ and find the capacitor you would need for the same $0.5\,\mathrm{V}$ requirement.
* Build it and measure the ripple with the capacitor you chose in [Task A4](Task_A4_5_1). By how much does the half-wave circuit miss the specification?
* The bridge costs you one extra diode drop and three extra diodes. State the design condition under which you would still choose the single diode.
````
