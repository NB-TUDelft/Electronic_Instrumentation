(Task_B1_5_1)=
#### Task B1: Decide whether a diode conducts
```{admonition} Estimated time: 10 min
:class: estimated-time no-content
```

Label the **anode** and **cathode** before you analyse a diode. Conventional forward current flows from anode to cathode. Choose a model:

* **Ideal:** a conducting diode has zero voltage across it; a blocking diode is an open circuit.
* **Constant drop:** a conducting silicon diode has $U_A-U_C\approx0.7\,\mathrm{V}$; a blocking diode is an open circuit.
* **Real:** forward voltage depends on current and temperature. You use the 1N4148 model in LTspice.

Use two questions to test the blocking assumption:

1. What are $U_A$ and $U_C$ if the diode blocks?
2. Does $U_A-U_C$ forward bias it under your chosen model?

If it does, use the conducting model and solve again. Check that the resulting current flows from anode to cathode.

##### Worked example: the series clipper

```{figure} images/EIschematics-diode2.svg
---
name: fig-series-clipper
height: 310px
---
The series clipper has its cathode at source `V1`, its anode at `V_out`, and resistor `R1` from the output to ground.
```

Use $R_1=1\,\mathrm{k}\Omega$ and

$$
U_{in}(t)=1\,\mathrm{V}\sin(2\pi\cdot1\,\mathrm{kHz}\cdot t).
$$

The amplitude is $1\,\mathrm{V}$, so the source swings from $-1\,\mathrm{V}$ to $+1\,\mathrm{V}$. This is $2\,\mathrm{V}$ peak to peak with a $1\,\mathrm{ms}$ period. In LTspice, enter `SINE(0 1 1k)`: offset, amplitude, frequency. Neglect the scope loading because its input resistance is much larger than $R_1$.

**Test the blocking assumption.** No current flows through $R_1$, so $U_{out}=0\,\mathrm{V}$. Thus $U_A=0\,\mathrm{V}$ and $U_C=U_{in}$.

**Ideal model.** The diode conducts when $-U_{in}>0$. Its voltage drop is then zero:

$$
U_{out}=
\begin{cases}
0\,\mathrm{V}, & U_{in}\geq0\,\mathrm{V},\\
U_{in}, & U_{in}<0\,\mathrm{V}.
\end{cases}
$$

**Constant-drop model.** Conduction starts near $U_{in}=-0.7\,\mathrm{V}$. While conducting, $U_{out}-U_{in}=0.7\,\mathrm{V}$:

$$
U_{out}\approx
\begin{cases}
0\,\mathrm{V}, & U_{in}\geq-0.7\,\mathrm{V},\\
U_{in}+0.7\,\mathrm{V}, & U_{in}<-0.7\,\mathrm{V}.
\end{cases}
$$

Sketch both predictions over one period. Mark the zero line, time axis, and voltage scale. Label the predicted negative peaks: $-1.0\,\mathrm{V}$ and $-0.3\,\mathrm{V}$. The model choice matters because the diode drop is a large fraction of the input amplitude.

### Anticipate

Predict each result before opening LTspice. Label every diode anode and cathode on your own copy of the schematic. A waveform sketch must show the zero line, both voltage axes, and at least one full period.

```{admonition} Hint for step 6
:class: hint dropdown
Estimate $I_D=(U_{in,p}-U_D)/R_1$ using $U_D=0.7\,\mathrm{V}$. Use the supplied curve values to revise $U_D$, then recalculate the current once. Predict whether increasing the source amplitude raises or lowers the clipping level.
```

```{admonition} Check your answer
:class: answer dropdown
With the anode at the output node, this diode limits the **positive** side of the output. The ideal model clips at $0\,\mathrm{V}$:

$$
U_{out}\approx
\begin{cases}
U_{in}, & U_{in}\leq 0\,\mathrm{V},\\
0\,\mathrm{V}, & U_{in}> 0\,\mathrm{V},
\end{cases}
$$

and the constant-drop model moves the clipping level up to $0.7\,\mathrm{V}$:

$$
U_{out}\approx
\begin{cases}
U_{in}, & U_{in}\leq 0.7\,\mathrm{V},\\
0.7\,\mathrm{V}, & U_{in}> 0.7\,\mathrm{V}.
\end{cases}
$$

Below the clipping level the diode blocks, no current flows in $R_1$, and the output follows the input. Above it the diode holds the output node and $R_1$ takes the difference, which is what sets the current:

$$
I_D\approx\frac{1.0\,\mathrm{V}-0.7\,\mathrm{V}}{1\,\mathrm{k}\Omega}=0.3\,\mathrm{mA}.
$$

The supplied real-model values suggest a clipping level near $0.55$ to $0.6\,\mathrm{V}$ at $1\,\mathrm{V}$ amplitude. At $5\,\mathrm{V}$ amplitude, expect a level near $0.75\,\mathrm{V}$ and a peak current around $4.3\,\mathrm{mA}$. Check these estimates in simulation. The forward drop changes with current.
```

```{admonition} Check your answer
:class: answer dropdown
The divider has

$$
U_{th}=1\,\mathrm{V}\cdot\frac{R_2}{R_1+R_2}=0.5\,\mathrm{V},
\qquad
R_{th}=R_1\parallel R_2=500\,\Omega.
$$

The anode is on the source side and the cathode at the output, so the diode begins to conduct near $U_{in}=0.5\,\mathrm{V}+0.7\,\mathrm{V}=1.2\,\mathrm{V}$:

$$
U_{out}\approx
\begin{cases}
0.5\,\mathrm{V}, & U_{in}\leq1.2\,\mathrm{V},\\
U_{in}-0.7\,\mathrm{V}, & U_{in}>1.2\,\mathrm{V}.
\end{cases}
$$

At the positive input peak, $U_{out}\approx7.3\,\mathrm{V}$ and

$$
I_D\approx\frac{7.3\,\mathrm{V}-0.5\,\mathrm{V}}{500\,\Omega}=13.6\,\mathrm{mA}.
$$

The predicted current is over forty times the value in [Task A1](Task_A1_5_1). ````
```

(Task_A4_5_1)=
#### Task A4: Choose a smoothing capacitor
```{admonition} Estimated time: 10 min
:class: estimated-time no-content
```

A capacitor across $R_{load}$ charges near each rectified peak and supplies the load between peaks. The result is a mean output with ripple, as shown in {numref}`fig-bridge-ripple`.

```{figure} images/bridge-ripple.svg
---
name: fig-bridge-ripple
height: 300px
---
Peak-to-peak ripple is the voltage the load loses between two capacitor recharge events.
```

Now put a real load on the bridge. Use the **silicon** bridge from [Task A3](Task_A3_5_1) with $R_{load}=1\,\mathrm{k}\Omega$ across the output terminals, as in {numref}`fig-bridge-loaded`, at $f=100\,\mathrm{Hz}$ and a predicted rectified peak of $8.6\,\mathrm{V}$. The design requirement is $U_{ripple,pp}\leq0.5\,\mathrm{V}$.

```{figure} images/EIschematics-fullbridge_redrawn.svg
---
name: fig-bridge-loaded
height: 360px
---
The same full bridge as in Task A3, redrawn with the load resistor $R_{load}$ across the output terminals.
```

For a full-wave rectifier, estimate

$$
U_{ripple,pp}\approx\frac{I_L}{2fC}.
$$

1. Use $I_L\approx8.6\,\mathrm{V}/R_{load}$ to calculate the minimum capacitance.
2. Choose the smallest value that meets the requirement from $1\,\mathrm{\mu F}$, $10\,\mathrm{\mu F}$, $47\,\mathrm{\mu F}$, $100\,\mathrm{\mu F}$, and $220\,\mathrm{\mu F}$.
3. Predict what happens to the ripple if $R_{load}$ is halved.
4. Predict what happens to the ripple if the source frequency is halved.
5. State two assumptions behind the ripple estimate.

```{admonition} Check your answer
:class: answer dropdown
The estimated load current is $8.6\,\mathrm{mA}$, so

$$
C_{min}\approx\frac{8.6\,\mathrm{mA}}{2(100\,\mathrm{Hz})(0.5\,\mathrm{V})}=86\,\mathrm{\mu F}.
$$

Choose $100\,\mathrm{\mu F}$, which predicts $0.43\,\mathrm{V}$ of ripple. The next value down, $47\,\mathrm{\mu F}$, predicts $0.91\,\mathrm{V}$ and fails. Halving $R_{load}$ roughly doubles the load current and therefore the ripple. Halving the frequency also doubles the ripple.

The estimate assumes that the load current is nearly constant between two short recharge pulses, that the ripple is small compared with the output voltage, and that the capacitor is recharged all the way to the peak each time. The last assumption needs a source and diode resistance that are small compared with $R_{load}$.
```

### Simulate

Use the built-in 1N4148 model for every silicon diode. A real diode does not switch at one exact voltage, so to compare circuits consistently, define the simulated transition as the point where $|I_D|$ first exceeds $0.1\,\mathrm{mA}$.

```{attention}
A bridge rectifier has separate source and load node pairs. In LTspice, ground one source terminal and leave both load terminals floating. Plot the differential load voltage as $V(out\_plus,out\_minus)$. Do not connect a second ground to the negative load terminal, because that shorts one bridge diode.
```

(Task_S1_5_1)=
#### Task S1: Simulate the shunt clipper and the biased clipper
```{admonition} Estimated time: 15 min
:class: estimated-time no-content
```

Build both circuits in one LTspice file:

1. the shunt clipper of [Task A1](Task_A1_5_1), with $R_1=1\,\mathrm{k}\Omega$, $R_{load}=1\,\mathrm{M}\Omega$ from the output node to ground to stand in for the scope input, and `SINE(0 1 1k)`;
2. the biased clipper of [Task A2](Task_A2_5_1), with $R_1=R_2=1\,\mathrm{k}\Omega$, a $1\,\mathrm{V}$ DC source at the top of the divider, and `SINE(0 8 1k)`.

Check the diode orientation in both schematics against {numref}`fig-shunt-clipper` and {numref}`fig-biased-clipper` before you run anything. A reversed diode still simulates, it just answers a different question.

Run a transient simulation over at least five periods with a maximum time step of $5\,\mathrm{\mu s}$.

For each circuit:

1. Plot $U_{in}$, $U_{out}$, and $I_D$.
2. Measure the minimum and the maximum $U_{out}$.
3. Measure the input voltage at which $|I_D|$ first exceeds $0.1\,\mathrm{mA}$.
4. Measure the peak diode current and the diode voltage at that current.
5. Explain one difference between the 1N4148 result and your ideal or constant-drop prediction.

Then change the shunt clipper source amplitude to $5\,\mathrm{V}$ and rerun. Record the new clipping level and the new peak diode current, and check them against your prediction in step 6 of [Task A1](Task_A1_5_1).

Save one labelled plot per circuit. Record the clipping levels and peak currents beside your predictions. Reuse those records at the bench and in [Task C1](Task_C1_5_1).

(Task_S2_5_1)=
#### Task S2: Simulate the silicon bridge and the capacitor design
```{admonition} Estimated time: 15 min
:class: estimated-time no-content
```

Build the bridge of {numref}`fig-bridge-loaded` with four 1N4148 diodes, `SINE(0 10 100)`, and $R_{load}=1\,\mathrm{k}\Omega$.

1. Simulate without a capacitor for at least $100\,\mathrm{ms}$.
2. Plot the source voltage and $V(out\_plus,out\_minus)$.
3. Measure the rectified output frequency and peak voltage, and compare them with your prediction from [Task A3](Task_A3_5_1).
4. Plot all four diode currents and confirm that the diagonal pairs alternate.
5. Add the capacitor you selected in [Task A4](Task_A4_5_1) across $R_{load}$.
6. Simulate for at least $300\,\mathrm{ms}$, then measure the mean output and the peak-to-peak ripple over the final two cycles.
7. Decide whether the result meets the $0.5\,\mathrm{V}$ requirement.

If the design misses the requirement, revise one assumption or one component value and run again. Record both results and explain the change. Save the schematic and the plots for the practicum.

## Practicum

### Implement and Investigate

Build the three clippers and one bridge using the ALPACA Playground LEDs. Use the **bench function generator** for every circuit. Compare the LED bridge with your silicon-bridge simulation from [Task S2](Task_S2_5_1). The smoothing-capacitor design stays in simulation.

Follow the pair roles in How To Work. Before switching on, one student traces each current path on the schematic while the other checks the connections. Swap roles for the next task.

```{attention}
Use the function generator return as the circuit reference. Both oscilloscope ground clips connect to that same node. For the bridge, put the two probe tips on `out_plus` and `out_minus` and use CH1 minus CH2. Both load terminals remain separate from ground. A ground clip on `out_minus` would bypass a bridge branch through the instruments' shared ground.
```

(Task_I1_5_1)=
#### Task I1: Build and reconcile the series clipper
```{admonition} Estimated time: 15 min
:class: estimated-time no-content
```

Build the series clipper of {numref}`fig-series-clipper` on the ALPACA breadboard with a 1N4148 and $R_1=1\,\mathrm{k}\Omega$. Work from the schematic, not from a breadboard layout. Drive the circuit from the **bench function generator**, not from the ALPACA DAC, with a $1\,\mathrm{kHz}$ sine at $1\,\mathrm{V}$ amplitude.

1. Identify the anode and the cathode of the diode before you insert it, and check the orientation with the DMM diode-test mode or with the ALPACA component tester. In this circuit the cathode faces the generator.
2. Put channel 1 on $U_{in}$ and channel 2 on $U_{out}$, both DC coupled. The predicted output peak is only a few hundred millivolts, so give channel 2 its own vertical scale and keep its zero line on screen.
3. Measure the negative peak of $U_{out}$. Compare it with both predictions of [Task B1](Task_B1_5_1), ideal and constant drop.
4. Measure the input voltage at which $U_{out}$ leaves $0\,\mathrm{V}$. That is the conduction threshold.
5. Save one labelled screenshot showing $U_{in}$ and $U_{out}$ together.

Both models of [Task B1](Task_B1_5_1) predicted the same output shape and disagreed about its size. Write down which of the three models you need to land within $50\,\mathrm{mV}$ of your measured peak, and what fraction of the input amplitude the diode drop took away.

(Task_I2_5_1)=
#### Task I2: Build and reconcile the shunt clipper
```{admonition} Estimated time: 20 min
:class: estimated-time no-content
```

Swap the two parts around to get the shunt clipper of {numref}`fig-shunt-clipper`: $R_1$ in series with the generator, and the 1N4148 from the output node to ground. The scope input is the load. Keep the generator at $1\,\mathrm{kHz}$ and $1\,\mathrm{V}$ amplitude.

1. Check the diode orientation again before you insert it. Here the anode faces the output node, so the part goes in the other way round than in [Task I1](Task_I1_5_1).
2. Put channel 1 on $U_{in}$ and channel 2 on $U_{out}$, both DC coupled.
3. Measure the clipping level, that is the flat value of $U_{out}$ at the positive peak.
4. Measure the input voltage at which $U_{out}$ departs from $U_{in}$.
5. Raise the amplitude to $5\,\mathrm{V}$ and measure the clipping level again.
6. Save one labelled screenshot per amplitude, showing $U_{in}$ and $U_{out}$ together.

The clipping level moved when you changed the amplitude, so the same part gave you two different forward drops. Record both clipping levels next to the two peak currents you calculated in [Task A1](Task_A1_5_1). If your measurement disagrees with [Task S1](Task_S1_5_1), change one model assumption at a time and note which change closes the gap.

(Task_I3_5_1)=
#### Task I3: Build and reconcile the biased clipper
```{admonition} Estimated time: 20 min
:class: estimated-time no-content
```

Build the biased clipper of {numref}`fig-biased-clipper` with a 1N4148 and $R_1=R_2=1\,\mathrm{k}\Omega$. The divider runs from the $1\,\mathrm{V}$ DC source `V2`, so set one channel of the bench supply to $1.00\,\mathrm{V}$ and check it with the DMM before you wire it in. Use the value you measured, not the value on the dial, in every comparison below. Drive the input from the function generator with a $1\,\mathrm{kHz}$ sine at $8\,\mathrm{V}$ amplitude.

```{attention}
Check the diode orientation once more. Here the anode faces the generator and the cathode faces the output node, which is the other way round than in [Task I2](Task_I2_5_1). The generator and the bench supply share one ground and nothing else, so keep the two returns separate everywhere except at that single ground point.
```

1. With the generator still disconnected, measure $U_{out}$ with the DMM. It should read the Thevenin voltage you calculated in [Task A2](Task_A2_5_1). If it does not, find the wiring error now rather than after the scope is connected.
2. Connect the generator. Put channel 1 on $U_{in}$ and channel 2 on $U_{out}$, both DC coupled.
3. Measure the input voltage at which $U_{out}$ leaves its blocking-state value, and compare it with the threshold you predicted.
4. Measure the peak $U_{out}$. Subtract it from the input peak to get the diode drop at the peak current, then use $I_D=(U_{out}-U_{th})/R_{th}$ to find the current that drop belongs to.
5. Save one labelled screenshot and compare it with your LTspice plot from [Task S1](Task_S1_5_1).

You have now measured the same 1N4148 at two currents that differ by more than a factor of forty: a fraction of a milliampere in [Task I2](Task_I2_5_1), and over ten milliamperes here. Write both drops down next to both currents.

(Task_I4_5_1)=
#### Task I4: Build and measure the ALPACA LED bridge
```{admonition} Estimated time: 35 min
:class: estimated-time no-content
```

Build one bridge from the four LED branches in the **Playground** area of your ALPACA, with $R_{load}=1\,\mathrm{k}\Omega$ across its output. Each branch contains an LED and an internal $390\,\Omega$ resistor. Both ends are available on header `J101`; neither end is tied to ALPACA ground.

```{figure} ../../images/LEDs_ALPACA.svg
---
name: fig-alpaca-leds-5-1
height: 400px
---
The four Playground LED branches and header `J101`; `_A` and `_C` identify the anode and cathode ends of each branch.
```

**Map and check the four nodes.** Use {numref}`fig-bridge-rectifier` and the header pinout in {numref}`fig-alpaca-leds-5-1` to draw your own schematic. Include the load, the internal resistors, and every header pin number. Assign the branches to $D_1$ through $D_4$ as in [Task A3](Task_A3_5_1).

```{admonition} Hint
:class: hint dropdown
Each branch joins one source node to one load node. First label the four nodes, then use the `_A` and `_C` pins to choose each branch's orientation.
```

One student wires the bridge while the other checks each branch against the drawing. Before switching on, confirm that the two source nodes and the two load nodes are four distinct nodes.

```{attention}
The internal resistors provide current limiting. Keep both load terminals separate from ALPACA ground and the instrument ground clips. Switch the generator output off before changing connections.
```

**Watch the current paths.** Set the generator to a sine at about $2\,\mathrm{Hz}$ and $10\,\mathrm{V}$ amplitude. Check the actual source amplitude with the scope.

1. Predict which pair lights in each half-cycle, then switch on and check. Trace the complete source, LED, load, LED, and return path for each pair.
2. Explain why the current through $R_{load}$ keeps the same direction when the source polarity reverses.

**Measure the floating output.** Swap roles and set the source to $100\,\mathrm{Hz}$ at the same amplitude. Put CH1 on `out_plus` and CH2 on `out_minus`, both DC coupled. Connect both ground clips to the generator return and display CH1 minus CH2.

3. Save one labelled capture showing both channels and their difference. Check the load polarity and count the rectified pulses per input period.
4. Measure the output peak in each half-cycle. Compare with the silicon-bridge prediction in [Task A3](Task_A3_5_1) and simulation in [Task S2](Task_S2_5_1). Explain how the LED forward drops and the two conducting $390\,\Omega$ resistors change the peaks. Different LED colours can give different peak heights.
5. Mark both probe tips and both ground clips on your schematic. Identify the branch that a ground clip on `out_minus` would bypass. Explain this on paper without making that connection.

### Compare and Conclude

Work in groups of four at your table. Use your existing predictions, plots, and measurements for the comparison below and your onepager. Agree on your explanations before asking the TA for check-off.

(Task_C1_5_1)=
#### Task C1: Reconcile prediction, simulation, and measurement
```{admonition} Estimated time: 15 min
:class: estimated-time no-content
```

```{admonition} Consider Tasks [B1](Task_B1_5_1), [S1](Task_S1_5_1), [S2](Task_S2_5_1), and [I1](Task_I1_5_1) to [I4](Task_I4_5_1).
:class: note no-content
```

Use your existing records to complete this table. Compare the same quantity in each row, such as a clipping level or an output peak. Name the assumption that best explains each mismatch.

| circuit | predicted feature | simulated feature | measured feature | explanation of mismatch |
|---|---|---|---|---|
| series clipper | | not simulated | | |
| shunt clipper, $1\,\mathrm{V}$ amplitude | | | | |
| shunt clipper, $5\,\mathrm{V}$ amplitude | | | | |
| biased clipper | | | | |
| silicon bridge | | | not built | |
| ALPACA LED bridge | load polarity and pulse count | not simulated | | |

Agree on one answer to each question:

1. For the clippers, where does the $0.7\,\mathrm{V}$ model agree with measurement to within $50\,\mathrm{mV}$? Use your measured drops and inferred currents as evidence.
2. The series and shunt clippers use the same two components. Explain why changing their positions changes the output shape.
3. Compare the silicon-bridge simulation with the LED-bridge measurement. Which differences come from the devices and series resistors? Which observations confirm the shared current paths?

(Task_C2_5_1)=
#### Task C2: TA check-off
```{admonition} Estimated time: 10 min
:class: estimated-time no-content
```

```{admonition} Consider Tasks [A4](Task_A4_5_1), [S2](Task_S2_5_1), [I2](Task_I2_5_1), [I3](Task_I3_5_1), and [I4](Task_I4_5_1).
:class: note no-content
```

At your table, show your evidence and explain the following to your TA:

1. Trace both LED-bridge current paths on your schematic. Explain the fixed load polarity and show where both scope ground clips connect.
2. Show whether the capacitor chosen in [Task A4](Task_A4_5_1) meets the $0.5\,\mathrm{V}$ ripple specification in your silicon-bridge simulation from [Task S2](Task_S2_5_1).
3. Use one measured diode drop and current from the clippers to explain when the constant-drop model is adequate.

````{admonition} Deep dive: diagnose one reversed bridge element
:class: deep-dive dropdown

A bridge with one element inserted backwards is one of the most common wiring faults in this practicum, and it does not look obviously broken on the scope.

Predict and simulate first. Take the silicon bridge of [Task S2](Task_S2_5_1) and reverse $D_1$ only.

* Redo the two-question analysis for both half-cycles. Which half-cycle still reaches the load, and which does not?
* There is now a path from one source node through $D_4$ and the reversed $D_1$ straight back to the other source node, without ever touching the load. Say what limits the current in that path, and what it does to the output peak.
* Predict the output waveform, then run the simulation and check the prediction.

Then reproduce it at the bench on the **LED** bridge of [Task I4](Task_I4_5_1), not on the silicon one. Every LED branch has its own internal $390\,\Omega$, so the fault path is current limited by construction.

* Reverse $D_1$, run the generator at about $2\,\mathrm{Hz}$, and record which LEDs light and when.
* Measure the output and compare it with your simulated prediction.
* Now say what the same fault would do to the silicon bridge of [Task S2](Task_S2_5_1), where nothing limits the fault path except the generator output impedance. Do not build that version.
* Write down the one scope observation that would let you spot this fault in under a minute next time.
````
