#### Task B1: Decide whether a diode conducts
```{admonition} Estimated time: 15 min
:class: estimated-time no-content
```

A diode has an **anode** and a **cathode**. Conventional current can flow from anode to cathode when the diode is forward biased. You choose a model before solving the circuit.

* **Ideal model:** a conducting diode has zero voltage across it. A blocking diode is an open circuit.
* **Constant-drop model:** a conducting silicon diode has $U_A - U_C \approx 0.7\,\mathrm{V}$. A blocking diode is an open circuit.
* **Real model:** current changes smoothly with voltage. The forward voltage depends on current and temperature. LTspice represents this behaviour with a component model.

```{attention}
A conducting diode is not a wire when you use the constant-drop model. Keep the $0.7\,\mathrm{V}$ drop in every voltage equation.
```

Use the following two questions whenever you analyze a diode circuit:

1. What are $U_A$ and $U_C$ if the diode is blocking?
2. Does $U_A-U_C$ forward bias the diode under your chosen model?

If the answer to the second question is yes, replace the diode by its conducting model and solve the circuit again.

Consider the half-wave rectifier in {numref}`fig-half-wave-rectifier`. Use

$$
U_{in}(t)=1\,\mathrm{V}\sin(2\pi\cdot 1\,\mathrm{kHz}\cdot t)
$$

and $R_L=1\,\mathrm{k}\Omega$.

```{figure} images/half-wave-rectifier.svg
---
name: fig-half-wave-rectifier
height: 300px
---
A series diode passes one part of the input cycle to the load.
```

For an ideal diode, the diode blocks when $U_{in}\leq 0\,\mathrm{V}$. The load then pulls $U_{out}$ to $0\,\mathrm{V}$. The diode conducts when $U_{in}>0\,\mathrm{V}$, so $U_{out}=U_{in}$.

For the constant-drop model, the result is

$$
U_{out}\approx
\begin{cases}
0, & U_{in}\leq 0.7\,\mathrm{V},\\
U_{in}-0.7\,\mathrm{V}, & U_{in}>0.7\,\mathrm{V}.
\end{cases}
$$

At the positive input peak, this model predicts $U_{out}\approx 0.3\,\mathrm{V}$ and

$$
I_D\approx\frac{0.3\,\mathrm{V}}{1\,\mathrm{k}\Omega}=0.3\,\mathrm{mA}.
$$

In your notepad, reproduce the two-question method for $U_{in}=-1.0\,\mathrm{V}$, $0.5\,\mathrm{V}$, and $1.0\,\mathrm{V}$. Record $U_{out}$ for both the ideal model and the constant-drop model. You will test these assumptions in [Task S1](Task_S1_5_1).

### Anticipate

Predict each result before opening LTspice. Label every diode anode and cathode on your own copy of the schematic. A waveform sketch must show the zero line, both voltage axes, and at least one full period.

(Task_A1_5_1)=
#### Task A1: Predict the shunt clipper
```{admonition} Estimated time: 10 min
:class: estimated-time no-content
```

#### Task A4: Choose a smoothing capacitor
```{admonition} Estimated time: 10 min
:class: estimated-time no-content
```

A capacitor across $R_L$ charges near each rectified peak and supplies the load between peaks. The result is a mean output with ripple, as shown in {numref}`fig-bridge-ripple`.

```{figure} images/bridge-ripple.svg
---
name: fig-bridge-ripple
height: 300px
---
Peak-to-peak ripple is the voltage lost between capacitor recharge events.
```

Use the bridge from [Task A3](Task_A3_5_1). The source frequency is $f=100\,\mathrm{Hz}$, $R_L=1\,\mathrm{k}\Omega$, and the predicted rectified peak is $3.6\,\mathrm{V}$. The maximum allowed ripple is $U_{ripple,pp}=0.4\,\mathrm{V}$.

For a full-wave rectifier, estimate

$$
U_{ripple,pp}\approx\frac{I_L}{2fC}.
$$

1. Use $I_L\approx3.6\,\mathrm{V}/R_L$ to calculate the minimum capacitance.
2. Choose the smallest expected value that meets the requirement from $1\,\mathrm{\mu F}$, $10\,\mathrm{\mu F}$, $47\,\mathrm{\mu F}$, and $100\,\mathrm{\mu F}$.
3. Predict what happens to the ripple if $R_L$ is halved.
4. Predict what happens to the ripple if the source frequency is halved.
5. State two assumptions behind the ripple estimate.

```{admonition} Check your answer
:class: answer dropdown
The estimated load current is $3.6\,\mathrm{mA}$. Therefore,

$$
C_{min}\approx\frac{3.6\,\mathrm{mA}}{2(100\,\mathrm{Hz})(0.4\,\mathrm{V})}=45\,\mathrm{\mu F}.
$$

Choose $47\,\mathrm{\mu F}$. Halving $R_L$ approximately doubles the load current and ripple. Halving the frequency also doubles the ripple.

The estimate assumes that the load current is nearly constant between short recharge pulses and that the ripple is small compared with the output voltage. Using the peak voltage to estimate the load current also treats the result as a conservative first design.
```

### Simulate

Use the built-in 1N4148 model for every diode. A real diode does not switch at one exact voltage. To compare circuits consistently, define the simulated transition as the point where $|I_D|$ first exceeds $0.1\,\mathrm{mA}$.

```{attention}
A bridge rectifier has separate source and load node pairs. In LTspice, ground one source terminal and leave both load terminals floating. Plot the differential load voltage as $V(out\_plus,out\_minus)$. Do not connect a second ground to the negative load terminal.
```
