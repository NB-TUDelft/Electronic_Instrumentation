#### Task B1: Decide whether a diode conducts
```{admonition} Estimated time: 15 min
:class: estimated-time no-content
```

A diode has an **anode** and a **cathode**. Conventional current flows from anode to cathode when the diode is forward biased, and the diode blocks current in the other direction. Before you can solve a diode circuit you have to choose a model for that behaviour.

* **Ideal model:** a conducting diode has zero voltage across it. A blocking diode is an open circuit. Conducts for any current flowing from anode to cathode.
* **Constant-drop model:** a conducting silicon diode has $U_A - U_C \approx 0.7\,\mathrm{V}$. A blocking diode is an open circuit.
* **Real model:** the current rises roughly exponentially with the forward voltage, so the forward voltage depends on the current and on the temperature. LTspice uses this model through a component model such as the 1N4148.

```{attention}
A conducting diode is not a wire when you use the constant-drop model. Keep the $0.7\,\mathrm{V}$ in every voltage equation.
```

Use these two questions whenever you meet a diode:

1. What are $U_A$ and $U_C$ if the diode blocks?
2. Does $U_A - U_C$ forward bias the diode under your chosen model?

If the answer to the second question is yes, replace the diode by its conducting model and solve the circuit again.

$$
U_{in}(t)=1\,\mathrm{V}\sin(2\pi\cdot 1\,\mathrm{kHz}\cdot t).
$$

```{admonition} How to read a source expression like this one
:class: tip dropdown
Read $U_{in}(t)=1\,\mathrm{V}\sin(2\pi\cdot 1\,\mathrm{kHz}\cdot t)$ from the outside in.

* The $(t)$ on the left says the value is a function of time. Every prediction about this circuit is therefore a prediction about a waveform, not about one number.
* $\sin(\ldots)$ runs between $-1$ and $+1$. Nothing is added to it, so the source is symmetric about $0\,\mathrm{V}$ and passes through zero going upwards at $t=0$.
* The $1\,\mathrm{V}$ in front is the **amplitude**, measured from zero to one peak. The input swings between $-1\,\mathrm{V}$ and $+1\,\mathrm{V}$, which is $2\,\mathrm{V}$ peak to peak. Function generators are usually set in peak to peak volts, so read the front panel carefully before you turn the knob.
* Inside the sine, the factor multiplying $t$ is the angular frequency $\omega = 2\pi f$. Here $f = 1\,\mathrm{kHz}$, so the period is $T = 1/f = 1\,\mathrm{ms}$. The $2\pi$ is there because a sine needs $2\pi$ radians for one full cycle.
* Two numbers follow from the amplitude: $U_{pp} = 2\,\mathrm{V}$, and $U_{rms} = 1\,\mathrm{V}/\sqrt{2} = 0.71\,\mathrm{V}$ for a full sine. As soon as a diode clips part of the waveform, the output is no longer a sine and that $\sqrt{2}$ no longer applies.
* In LTspice the same source is `SINE(0 1 1k)`, which is offset, amplitude, frequency. The second number is an amplitude, so that source also swings between $-1\,\mathrm{V}$ and $+1\,\mathrm{V}$.


Read every source in this manual the same way. A $10\,\mathrm{V}$ amplitude at $100\,\mathrm{Hz}$, for example, swings between $-10\,\mathrm{V}$ and $+10\,\mathrm{V}$ with a period of $10\,\mathrm{ms}$.
```

**Step 1: solve it with the ideal model.**

1. Assume the diode blocks. Then no current flows in $R_1$, and a resistor without current has no voltage across it, so $U_{out}=0\,\mathrm{V}$. The anode is at the output, so $U_A=0\,\mathrm{V}$, and the cathode is at the source, so $U_C=U_{in}$.
2. An ideal diode is forward biased as soon as $U_A-U_C>0$, which here means $-U_{in}>0$, that is $U_{in}<0\,\mathrm{V}$.

So the blocking assumption holds only while $U_{in}\geq 0\,\mathrm{V}$. During the negative half of the cycle the diode conducts, and an ideal conducting diode has no voltage across it, so the output follows the input:

$$
U_{out}\approx
\begin{cases}
0\,\mathrm{V}, & U_{in}\geq 0\,\mathrm{V},\\
U_{in}, & U_{in}< 0\,\mathrm{V}.
\end{cases}
$$

**Step 2: solve the same circuit with the constant-drop model.**

1. Question 1 gives the same answer, because a blocking diode is still an open circuit: $U_A=0\,\mathrm{V}$ and $U_C=U_{in}$.
2. Question 2 changes. The diode now needs $U_A-U_C\geq0.7\,\mathrm{V}$, so it needs $U_{in}\leq-0.7\,\mathrm{V}$.

Once it conducts, the diode holds $U_{out}-U_{in}=0.7\,\mathrm{V}$, so

$$
U_{out}\approx
\begin{cases}
0\,\mathrm{V}, & U_{in}\geq -0.7\,\mathrm{V},\\
U_{in}+0.7\,\mathrm{V}, & U_{in}< -0.7\,\mathrm{V}.
\end{cases}
$$

That is the whole method: assume a state, check the bias, then solve the circuit that the assumption leaves you with.

Notice how much the choice of model costs here. The ideal model predicts an output peak of $-1.0\,\mathrm{V}$, the constant-drop model predicts $-0.3\,\mathrm{V}$. At a $1\,\mathrm{V}$ amplitude the $0.7\,\mathrm{V}$ is most of the signal, so the model is not a detail. It decides the answer.

In your notepad, sketch both predictions over one full period, with the zero line and both voltage axes marked.

:class: hint dropdown

```{admonition} Check your answer
:class: answer dropdown
With the anode at the output node, this diode limits the **positive** side of the output. The ideal model clips at $0\,\mathrm{V}$:

$$
U_{out}\approx
\begin{cases}
U_{in}, & U_{in}\leq 0\,\mathrm{V},\\
0\,\mathrm{V}, & U_{in}> 0\,\mathrm{V},
\end{cases}
$$

and the constant-drop model moves the clipping level up to $0.7\,\mathrm{V}$:

$$
U_{out}\approx
\begin{cases}
U_{in}, & U_{in}\leq 0.7\,\mathrm{V},\\
0.7\,\mathrm{V}, & U_{in}> 0.7\,\mathrm{V}.
\end{cases}
$$

Below the clipping level the diode blocks, no current flows in $R_1$, and the output follows the input. Above it the diode holds the output node and $R_1$ takes the difference, which is what sets the current:

$$
I_D\approx\frac{1.0\,\mathrm{V}-0.7\,\mathrm{V}}{1\,\mathrm{k}\Omega}=0.3\,\mathrm{mA}.
$$

For the real model, start from that $0.3\,\mathrm{mA}$. The curve gives about $0.55\,\mathrm{V}$ there, which raises the current to $(1.0\,\mathrm{V}-0.55\,\mathrm{V})/1\,\mathrm{k}\Omega=0.45\,\mathrm{mA}$, which pulls the drop back up a little. Expect a measured clipping level of about $0.55$ to $0.6\,\mathrm{V}$, with a peak current near $0.43\,\mathrm{mA}$. At this current the $0.7\,\mathrm{V}$ model is too pessimistic by more than $100\,\mathrm{mV}$.

At a $5\,\mathrm{V}$ amplitude the same loop gives a peak current of about $4.3\,\mathrm{mA}$, and the curve puts the drop near $0.75\,\mathrm{V}$. So the clipping level is not a fixed property of the diode. It follows the current, and here the source amplitude and $R_1$ are what set that current.

Note that at their peaks this circuit and the series clipper of [Task B1](Task_B1_5_1) carry the same $0.3\,\mathrm{mA}$ through the same two parts. Only the order changed, and with it the output shape.
```

````

(Task_A4_5_1)=
#### Task A4: Choose a smoothing capacitor
```{admonition} Estimated time: 10 min
:class: estimated-time no-content
```

A capacitor across $R_L$ charges near each rectified peak and supplies the load between peaks. The result is a mean output with ripple, as shown in {numref}`fig-bridge-ripple`.

```{figure} images/bridge-ripple.svg
---
name: fig-bridge-ripple
height: 300px
---
Peak-to-peak ripple is the voltage the load loses between two capacitor recharge events.
```

Use the **silicon** bridge from [Task A3](Task_A3_5_1): $f=100\,\mathrm{Hz}$, $R_L=1\,\mathrm{k}\Omega$, and a predicted rectified peak of $8.6\,\mathrm{V}$. The design requirement is $U_{ripple,pp}\leq0.5\,\mathrm{V}$.

For a full-wave rectifier, estimate

$$
U_{ripple,pp}\approx\frac{I_L}{2fC}.
$$

1. Use $I_L\approx8.6\,\mathrm{V}/R_L$ to calculate the minimum capacitance.
2. Choose the smallest value that meets the requirement from $1\,\mathrm{\mu F}$, $10\,\mathrm{\mu F}$, $47\,\mathrm{\mu F}$, $100\,\mathrm{\mu F}$, and $220\,\mathrm{\mu F}$.
3. Predict what happens to the ripple if $R_L$ is halved.
4. Predict what happens to the ripple if the source frequency is halved.
5. State two assumptions behind the ripple estimate.

```{admonition} Check your answer
:class: answer dropdown
The estimated load current is $8.6\,\mathrm{mA}$, so

$$
C_{min}\approx\frac{8.6\,\mathrm{mA}}{2(100\,\mathrm{Hz})(0.5\,\mathrm{V})}=86\,\mathrm{\mu F}.
$$

Choose $100\,\mathrm{\mu F}$, which predicts $0.43\,\mathrm{V}$ of ripple. The next value down, $47\,\mathrm{\mu F}$, predicts $0.91\,\mathrm{V}$ and fails. Halving $R_L$ roughly doubles the load current and therefore the ripple. Halving the frequency also doubles the ripple.

The estimate assumes that the load current is nearly constant between two short recharge pulses, that the ripple is small compared with the output voltage, and that the capacitor is recharged all the way to the peak each time. The last assumption needs a source and diode resistance that are small compared with $R_L$, which is exactly what fails in the LED bridge.
```

### Simulate

Use the built-in 1N4148 model for every silicon diode. A real diode does not switch at one exact voltage, so to compare circuits consistently, define the simulated transition as the point where $|I_D|$ first exceeds $0.1\,\mathrm{mA}$.

```{attention}
A bridge rectifier has separate source and load node pairs. In LTspice, ground one source terminal and leave both load terminals floating. Plot the differential load voltage as $V(out\_plus,out\_minus)$. Do not connect a second ground to the negative load terminal, because that shorts one bridge diode.
```

(Task_S1_5_1)=
#### Task S1: Simulate the shunt clipper and the biased clipper
```{admonition} Estimated time: 20 min
:class: estimated-time no-content
```

Build both circuits in one LTspice file:

1. the shunt clipper of [Task A1](Task_A1_5_1), with $R_1=1\,\mathrm{k}\Omega$, $R_L=1\,\mathrm{M}\Omega$ from the output node to ground to stand in for the scope input, and `SINE(0 1 1k)`;
2. the biased clipper of [Task A2](Task_A2_5_1), with $R_1=R_2=1\,\mathrm{k}\Omega$, a $1\,\mathrm{V}$ DC source at the top of the divider, and `SINE(0 8 1k)`.

Check the diode orientation in both schematics against {numref}`fig-shunt-clipper` and {numref}`fig-biased-clipper` before you run anything. A reversed diode still simulates, it just answers a different question.

Run a transient simulation over at least five periods with a maximum time step of $5\,\mathrm{\mu s}$.

For each circuit:

1. Plot $U_{in}$, $U_{out}$, and $I_D$.
2. Measure the minimum and the maximum $U_{out}$.
3. Measure the input voltage at which $|I_D|$ first exceeds $0.1\,\mathrm{mA}$.
4. Measure the peak diode current and the diode voltage at that current.
5. Explain one difference between the 1N4148 result and your ideal or constant-drop prediction.

Then change the shunt clipper source amplitude to $5\,\mathrm{V}$ and rerun. Record the new clipping level and the new peak diode current, and check them against your prediction in step 6 of [Task A1](Task_A1_5_1).

Complete this table in your notepad:

| circuit | ideal prediction | constant-drop prediction | 1N4148 simulation | assumption that explains the difference |
|---|---|---|---|---|
| shunt clipper, $1\,\mathrm{V}$ |  |  |  |  |
| shunt clipper, $5\,\mathrm{V}$ |  |  |  |  |
| biased clipper |  |  |  |  |

Save one clearly labelled plot per circuit. You compare them with your oscilloscope traces during the practicum.

(Task_S2_5_1)=
#### Task S2: Simulate the silicon bridge and the capacitor design
```{admonition} Estimated time: 15 min
:class: estimated-time no-content
```

Build the bridge from [Task A3](Task_A3_5_1) with four 1N4148 diodes, `SINE(0 10 100)`, and $R_L=1\,\mathrm{k}\Omega$.

1. Simulate without a capacitor for at least $100\,\mathrm{ms}$.
2. Plot the source voltage and $V(out\_plus,out\_minus)$.
3. Measure the rectified output frequency and peak voltage, and compare them with your prediction from [Task A3](Task_A3_5_1).
4. Plot all four diode currents and confirm that the diagonal pairs alternate.
5. Add the capacitor you selected in [Task A4](Task_A4_5_1) across $R_L$.
6. Simulate for at least $300\,\mathrm{ms}$, then measure the mean output and the peak-to-peak ripple over the final two cycles.
7. Decide whether the result meets the $0.5\,\mathrm{V}$ requirement.

If the design misses the requirement, revise one assumption or one component value and run again. Record both results and explain the change. Save the schematic and the plots for the practicum.

(Task_S3_5_1)=
#### Task S3: Simulate the ALPACA LED bridge
```{admonition} Estimated time: 10 min
:class: estimated-time no-content
```

Copy the bridge from [Task S2](Task_S2_5_1) and turn each arm into one ALPACA LED branch: a $390\,\Omega$ resistor in series with an LED. Keep `SINE(0 10 100)` and $R_L=1\,\mathrm{k}\Omega$, and remove the smoothing capacitor.

An LED is not in the standard LTspice diode library, so you supply the constant-drop model yourself. Add these lines as a SPICE directive and set the value of each diode symbol to the matching model name:

```text
.model LED_GRN D(Ron=1 Roff=1G Vfwd=2.2 epsilon=0.1)
.model LED_YEL D(Ron=1 Roff=1G Vfwd=2.0 epsilon=0.1)
.model LED_RED D(Ron=1 Roff=1G Vfwd=1.9 epsilon=0.1)
.model LED_BLU D(Ron=1 Roff=1G Vfwd=3.0 epsilon=0.1)
```

`Vfwd` is the forward threshold, `Ron` the on-resistance, and `epsilon` a small rounding of the knee that helps the solver converge. This is the constant-drop model of [Task B1](Task_B1_5_1), written out for the simulator. If your LTspice version rejects `Vfwd`, use an ordinary diode in series with a DC voltage source equal to $U_f$ instead.

1. Use the colour assignment you chose in [Task A3](Task_A3_5_1).
2. Plot $V(out\_plus,out\_minus)$ and the current in each of the four branches.
3. Measure the peak output voltage of each half-cycle separately. Compare both with your hand calculation.
4. Measure the conduction interval of one branch as a fraction of the half-period, and explain why it is shorter than for the silicon bridge.
5. Now add the $100\,\mathrm{\mu F}$ capacitor across $R_L$ and simulate again. Report the mean output voltage and the ripple, and explain why the mean sits so far below the peak you measured in step 3.

Save the two output plots, with and without the capacitor. You will compare them with the bench result in [Task I5](Task_I5_5_1).

## Practicum

### Implement and investigate

Work in pairs. At home you predicted four diode circuits and simulated two of them. At the bench you build all three clippers, the series clipper of [Task B1](Task_B1_5_1), the shunt clipper of [Task A1](Task_A1_5_1), and the biased clipper of [Task A2](Task_A2_5_1), and then the bridge rectifier of [Task A3](Task_A3_5_1) in both its silicon and its LED form. Everything runs from the **bench function generator**, and for each circuit you reconcile prediction, simulation, and measurement.

Divide the work concretely. For each task, one student builds the circuit on the ALPACA breadboard while the other configures the function generator and the oscilloscope. Swap roles before the next task. During the bridge tasks, one student traces and checks each node against the schematic while the other wires the next branch. Never energize a circuit until both of you agree on what is on the breadboard.

```{attention}
Connect **one** ground reference only, at the function generator negative terminal. Do not add a second ground wire from the scope to the negative load terminal of a bridge, because that short-circuits one bridge element and distorts the measurement. Measure a floating bridge output differentially, with one channel on each load terminal and the scope math subtract function.
```

(Task_I1_5_1)=
#### Task I1: Build and reconcile the series clipper
```{admonition} Estimated time: 15 min
:class: estimated-time no-content
```

Build the series clipper of {numref}`fig-series-clipper` on the ALPACA breadboard with a 1N4148 and $R_1=1\,\mathrm{k}\Omega$. Work from the schematic, not from a breadboard layout. Drive the circuit from the **bench function generator**, not from the ALPACA DAC, with a $1\,\mathrm{kHz}$ sine at $1\,\mathrm{V}$ amplitude.

1. Identify the anode and the cathode of the diode before you insert it, and check the orientation with the DMM diode-test mode or with the ALPACA component tester. In this circuit the cathode faces the generator.
2. Put channel 1 on $U_{in}$ and channel 2 on $U_{out}$, both DC coupled. The predicted output peak is only a few hundred millivolts, so give channel 2 its own vertical scale and keep its zero line on screen.
3. Measure the negative peak of $U_{out}$. Compare it with both predictions of [Task B1](Task_B1_5_1), ideal and constant drop.
4. Measure the input voltage at which $U_{out}$ leaves $0\,\mathrm{V}$. That is the conduction threshold.
5. Save one labelled screenshot showing $U_{in}$ and $U_{out}$ together.

Both models of [Task B1](Task_B1_5_1) predicted the same output shape and disagreed about its size. Write down which of the three models you need to land within $50\,\mathrm{mV}$ of your measured peak, and what fraction of the input amplitude the diode drop took away.

(Task_I2_5_1)=
#### Task I2: Build and reconcile the shunt clipper
```{admonition} Estimated time: 20 min
:class: estimated-time no-content
```

Swap the two parts around to get the shunt clipper of {numref}`fig-shunt-clipper`: $R_1$ in series with the generator, and the 1N4148 from the output node to ground. The scope input is the load. Keep the generator at $1\,\mathrm{kHz}$ and $1\,\mathrm{V}$ amplitude.

1. Check the diode orientation again before you insert it. Here the anode faces the output node, so the part goes in the other way round than in [Task I1](Task_I1_5_1).
2. Put channel 1 on $U_{in}$ and channel 2 on $U_{out}$, both DC coupled.
3. Measure the clipping level, that is the flat value of $U_{out}$ at the positive peak.
4. Measure the input voltage at which $U_{out}$ departs from $U_{in}$.
5. Raise the amplitude to $5\,\mathrm{V}$ and measure the clipping level again.
6. Save one labelled screenshot per amplitude, showing $U_{in}$ and $U_{out}$ together.

The clipping level moved when you changed the amplitude, so the same part gave you two different forward drops. Record both clipping levels next to the two peak currents you calculated in [Task A1](Task_A1_5_1). If your measurement disagrees with [Task S1](Task_S1_5_1), change one model assumption at a time and note which change closes the gap.

(Task_I3_5_1)=
#### Task I3: Build and reconcile the biased clipper
```{admonition} Estimated time: 20 min
:class: estimated-time no-content
```

Build the biased clipper of {numref}`fig-biased-clipper` with a 1N4148 and $R_1=R_2=1\,\mathrm{k}\Omega$. The divider runs from the $1\,\mathrm{V}$ DC source `V2`, so set one channel of the bench supply to $1.00\,\mathrm{V}$ and check it with the DMM before you wire it in. Use the value you measured, not the value on the dial, in every comparison below. Drive the input from the function generator with a $1\,\mathrm{kHz}$ sine at $8\,\mathrm{V}$ amplitude.

```{attention}
Check the diode orientation once more. Here the anode faces the generator and the cathode faces the output node, which is the other way round than in [Task I2](Task_I2_5_1). The generator and the bench supply share one ground and nothing else, so keep the two returns separate everywhere except at that single ground point.
```

1. With the generator still disconnected, measure $U_{out}$ with the DMM. It should read the Thevenin voltage you calculated in [Task A2](Task_A2_5_1). If it does not, find the wiring error now rather than after the scope is connected.
2. Connect the generator. Put channel 1 on $U_{in}$ and channel 2 on $U_{out}$, both DC coupled.
3. Measure the input voltage at which $U_{out}$ leaves its blocking-state value, and compare it with the threshold you predicted.
4. Measure the peak $U_{out}$. Subtract it from the input peak to get the diode drop at the peak current, then use $I_D=(U_{out}-U_{th})/R_{th}$ to find the current that drop belongs to.
5. Save one labelled screenshot and compare it with your LTspice plot from [Task S1](Task_S1_5_1).

You have now measured the same 1N4148 at two currents that differ by more than a factor of forty: a fraction of a milliampere in [Task I2](Task_I2_5_1), and over ten milliamperes here. Write both drops down next to both currents.

(Task_I4_5_1)=
#### Task I4: Build the silicon bridge and meet the ripple requirement
```{admonition} Estimated time: 40 min
:class: estimated-time no-content
```

Build the bridge of {numref}`fig-bridge-rectifier` from four 1N4148 diodes and $R_L=1\,\mathrm{k}\Omega$ on the breadboard. Leave the capacitor out for now.

Before you energize the circuit:

1. Check each of the four diodes with the DMM diode-test mode, and annotate your schematic with the anode and cathode you measured.
2. Confirm that the two source nodes and the two load nodes are four distinct nodes, and that no extra ground wire bridges them.

Drive the bridge from the function generator with a $100\,\mathrm{Hz}$ sine at $10\,\mathrm{V}$ amplitude, and measure $U_{out}=U(out\_plus)-U(out\_minus)$ differentially.

3. Measure the rectified output frequency. Compare it with the source frequency and with your prediction from [Task A3](Task_A3_5_1).
4. Measure the rectified peak voltage and compare it with the two-drop prediction.
5. Swap the two generator leads at the source nodes. Confirm that the load polarity does not reverse. That is the defining property of a full bridge.

Now add the capacitor you selected in [Task A4](Task_A4_5_1) across $R_L$.

```{attention}
If you use an electrolytic capacitor, observe its polarity. The positive terminal goes to `out_plus`. A reversed electrolytic capacitor can fail and vent electrolyte.
```

6. Measure the mean output voltage and the peak-to-peak ripple with the scope measurement functions.
7. Decide whether the result meets the $0.5\,\mathrm{V}$ requirement.
8. Repeat with one capacitor value lower, or with $R_L$ halved, and record the new ripple.

In your notepad, state whether the original design met the specification. If it did not, revise the value, name the assumption that caused the miss, and record the corrected result.

(Task_I5_5_1)=
#### Task I5: Make the bridge current paths visible with the ALPACA LEDs
```{admonition} Estimated time: 30 min
:class: estimated-time no-content
```

Now build the same bridge from the four Playground LED branches of your ALPACA, as in {numref}`fig-alpaca-led-bridge`, using the header pins listed in [Task A3](Task_A3_5_1). Keep $R_L=1\,\mathrm{k}\Omega$ on the breadboard.

```{attention}
Every LED branch already contains its own internal $390\,\Omega$ resistor, so you must not add one, and you must not leave it out of your calculations either. Both terminals of every branch are floating: neither is tied to the ALPACA ground. Do not connect `out_minus` to ground.
```

**Step 1: measure the forward voltage of each branch.** A diode-test reading is not usable here, because the internal $390\,\Omega$ and a forward voltage above $2\,\mathrm{V}$ push most meters over range. Instead, drive one branch from the $5\,\mathrm{V}$ rail through an external $R_s=470\,\Omega$ and measure your way to $U_f$:

1. Measure $U_{Rs}$ across the external resistor and calculate the branch current $I=U_{Rs}/R_s$.
2. Measure $U_{branch}$ across the whole branch, from the anode-side pin to the cathode pin.
3. Calculate $U_f=U_{branch}-(390\,\Omega)I$.

Record $U_f$ **together with the current it was measured at**, for all four colours. Compare the values with the typical numbers you used in [Task A3](Task_A3_5_1).

**Step 2: watch the two paths alternate.** Wire the bridge with the colour assignment you chose in [Task A3](Task_A3_5_1). Drive it from the function generator at $10\,\mathrm{V}$ amplitude and about $2\,\mathrm{Hz}$, slow enough to follow with your eyes.

4. State your prediction out loud before you switch the generator on: which two LEDs light together, and which two light on the other half-cycle?
5. Record which pair lights in each half-cycle and confirm that the diagonal pairs alternate.
6. Reverse the two generator leads and confirm that the same pairs still alternate.

**Step 3: measure the rectified output.** Set the generator back to $100\,\mathrm{Hz}$ at $10\,\mathrm{V}$ amplitude and measure $U_{out}$ differentially.

7. Measure the peak output voltage of **each** half-cycle. They differ. Explain the difference from your $U_f$ measurements.
8. Compare the peak output with the silicon bridge result of [Task I4](Task_I4_5_1) at the same source setting. Account for the missing volts, term by term: two forward drops and two internal $390\,\Omega$ resistors.
9. Add the $100\,\mathrm{\mu F}$ capacitor across $R_L$. Measure the mean output and the ripple, and explain why the mean sits well below the peak from step 7. Compare with [Task S3](Task_S3_5_1).

Your final schematic must show all four internal $390\,\Omega$ resistors and the header pin number of every connection.

### Compare and Conclude

Work in groups of four at your table. Pool your predictions, LTspice plots, and bench measurements, then fill the tables below in your notepad. The completed tables are the core of your onepager report. The goal is not the formulas, which you already wrote down, but the side-by-side view of prediction, simulation, and measurement.

(Task_C1_5_1)=
#### Task C1: Prediction, simulation, measurement matrix
```{admonition} Consider Tasks [B1](Task_B1_5_1), [A1](Task_A1_5_1), [A2](Task_A2_5_1), [A3](Task_A3_5_1), [S1](Task_S1_5_1), [S2](Task_S2_5_1), [S3](Task_S3_5_1), [I1](Task_I1_5_1), [I2](Task_I2_5_1), [I3](Task_I3_5_1), [I4](Task_I4_5_1), and [I5](Task_I5_5_1).
:class: note no-content
```

Fill one row per circuit. For each row, write the rule that decides whether a diode conducts, the key predicted feature, the simulated feature, the measured feature, and the single best explanation of any mismatch.

| circuit | conduction rule | predicted feature | simulated feature | measured feature | best explanation of mismatch |
|---|---|---|---|---|---|
| series clipper, $1\,\mathrm{V}$ |  |  |  |  |  |
| shunt clipper, $1\,\mathrm{V}$ |  |  |  |  |  |
| shunt clipper, $5\,\mathrm{V}$ |  |  |  |  |  |
| biased clipper |  |  |  |  |  |
| silicon bridge |  |  |  |  |  |
| ALPACA LED bridge |  |  |  |  |  |

One cell in that table stays empty on purpose, because you never simulated the series clipper. Write "not simulated" there rather than leaving a blank, and say in one line what that missing cell would have told you.

Then agree on one answer to each of these three questions, and write the answer your table settles on.

1. Across all six rows, at which currents did the $0.7\,\mathrm{V}$ model predict your measurement to within $50\,\mathrm{mV}$, and where did it fail?
2. The series clipper and the shunt clipper contain the same diode and the same resistor. Name what decides the shape of the output, and state what the two circuits nevertheless have in common at their peaks.
3. The LED bridge and the silicon bridge share a topology but not a waveform. Name the two separate device properties that account for the difference, and say which of the two costs you more output voltage.

(Task_C2_5_1)=
#### Task C2: TA check-off
```{admonition} Consider Tasks [A3](Task_A3_5_1), [A4](Task_A4_5_1), [I2](Task_I2_5_1), [I3](Task_I3_5_1), [I4](Task_I4_5_1), and [I5](Task_I5_5_1).
:class: note no-content
```

Before you leave, explain the following to your TA without looking at the finished wiring:

1. Trace one bridge current path from the source, through two elements, to the load, and back. State which two conduct and why the load polarity stays fixed.
2. Show the ripple design evidence: the capacitor you chose in [Task A4](Task_A4_5_1), the ripple you measured in [Task I4](Task_I4_5_1), and whether the design met the $0.5\,\mathrm{V}$ specification.
3. State one condition under which the constant $0.7\,\mathrm{V}$ model is adequate and one under which it is not, using the drops and the currents you recorded in [Task I2](Task_I2_5_1) and [Task I3](Task_I3_5_1).

````{admonition} Deep dive: diagnose one reversed bridge element
:class: deep-dive dropdown

A bridge with one element inserted backwards is one of the most common wiring faults in this practicum, and it does not look obviously broken on the scope.

Predict and simulate first. In the LED bridge of [Task S3](Task_S3_5_1), reverse $D_1$ only.

* Redo the two-question analysis for both half-cycles. Which half-cycle still reaches the load, and which does not?
* There is now a path from `in_b` through $D_4$ and the reversed $D_1$ back to `in_a` that never touches the load. Calculate the current in that path, and state what it does to the output peak.
* Predict the output waveform, and predict which LEDs light and when.

Then reproduce it at the bench. The LED bridge is the safe place to do this: every branch has its own internal $390\,\Omega$, so the fault path is current limited by construction.

* Reverse $D_1$, measure the output, and compare with your simulation.
* Now say what the same fault would do to the silicon bridge of [Task I4](Task_I4_5_1), where nothing limits the fault path except the generator output impedance. Do not build that version.
* Write down the one scope observation that would let you spot this fault in under a minute next time.
````
