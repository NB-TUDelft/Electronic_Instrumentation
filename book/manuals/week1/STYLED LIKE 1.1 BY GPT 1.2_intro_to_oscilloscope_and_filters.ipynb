- The input channels are where we connect the signal we want to observe.
- The display shows the measured waveform.
- The horizontal controls set the time scale and move the signal left or right.
- The vertical controls set the voltage scale and move the signal up or down.
- The trigger settings determine when the oscilloscope starts drawing the waveform on the screen.

#### RC filters

In this manual you will also build a simple RC high-pass filter. By varying the input frequency and measuring the output with the oscilloscope, you can observe how the filter passes high-frequency components more easily than low-frequency components.

#### Task A1: Effect of AC/DC coupling and signal frequency on the measured voltage
When you input a signal into the oscilloscope, there are two common coupling modes:

* AC coupling
* DC coupling

In DC coupling mode, the oscilloscope displays the full signal. In AC coupling mode, the average value is removed, so only the changing part of the signal remains visible.

How does this happen? In DC coupling mode, the input signal goes directly to the oscilloscope. In AC coupling mode, the signal first passes through a capacitor. The input therefore behaves like a simple high-pass RC filter: low-frequency components are reduced, and the DC component is blocked.
