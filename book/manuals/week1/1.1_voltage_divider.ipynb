
(Task_S1_1_1)=
#### Task S1: Simple voltage divider
In tasks A1 and A2, you conducted some calculations. Now it's time to simulate the circuit and verify your results. You can build the circuit in LTspice once, then adjust the resistors' values and simulate again. For the following resistor pairs, you can also measure current through the resistors at the same time, to avoid simulating the circuit again:
1. $R_1 = R_2 = 100\,\Omega$
2. $R_1 = R_2 = 1\,\mathrm{k}\Omega$
3. $R_1 = R_2 = 100\,\mathrm{k}\Omega$
4. $R_1 = R_2 = 10\,\mathrm{M}\Omega$

```{figure} images/voltage_divider.svg
---
name: Voltage_divider2
height: 400px
---
Voltage divider circuit
```
Record the simulated output voltages for the following resistor pairs:
1. $R_1 = 100\,\Omega$, $R_2 = 100\,\Omega$
2. $R_1 = 1\,\mathrm{k}\Omega$, $R_2 = 1\,\mathrm{k}\Omega$
3. $R_1 = 100\,\mathrm{k}\Omega$, $R_2 = 100\,\mathrm{k}\Omega$
4. $R_1 = 10\,\mathrm{M}\Omega$, $R_2 = 10\,\mathrm{M}\Omega$
5. $R_1 = 1\,\mathrm{k}\Omega$, $R_2 = 220\,\Omega$
6. $R_1 = 220\,\Omega$, $R_2 = 1\,\mathrm{k}\Omega$

Also record the simulated currents for the following resistor pairs:
1. $R_1 = R_2 = 100\,\Omega$
2. $R_1 = R_2 = 1\,\mathrm{k}\Omega$
3. $R_1 = R_2 = 100\,\mathrm{k}\Omega$
4. $R_1 = R_2 = 10\,\mathrm{M}\Omega$

Keep brief notes in your notepad about any simulation results that surprise you.

(Task_S2_1_1)=
#### Task S2: Use LTspice to build the simulation of connected voltage dividers
Simulate these circuits introduced in task A4, and check whether your output voltage calculations are correct.
To make this task quicker, build the basic dividers first, then connect them in three different ways.

```{figure} images/stacked_dividers.svg
---
height: 300px
name: fig-stacked-dividers2
---
Voltage dividers in series
```

```{figure} images/two_dividers_thevinen.svg
---
height: 300px
name: fig-two-dividers-thevinen2
---
Voltage dividers in parallel
```

## Practicum
### Implement and investigate
During the practicum, work in pairs and divide the practical work as equally as possible. If one student builds part of a circuit or performs one set of measurements, the other student should build the other part or perform the complementary measurements, and both students should stay involved in checking the results.
(Task_I1_1_1)=
#### Task I1: Building simple voltage dividers on a breadboard
Using the breadboard, cables and resistors, build a voltage divider circuit. Remember to connect the power rails of the breadboard to the power source. You can build circuits next to each other, but preferably go one by one, to ensure that the measurements are unbiased.

Divide the building work equally. For example, if you are preparing several divider configurations, one student can build the first half of the dividers and the other student can build the second half, while both students check each wiring step before measuring.

```{figure} images/voltage_divider.svg
---
name: Voltage_divider3
height: 400px
---
Voltage divider circuit
```

After building the circuit, measure the voltage at the output of the divider for the following six resistor pairs:
1. $R_1 = 100\,\Omega$, $R_2 = 100\,\Omega$
2. $R_1 = 1\,\mathrm{k}\Omega$, $R_2 = 1\,\mathrm{k}\Omega$
3. $R_1 = 100\,\mathrm{k}\Omega$, $R_2 = 100\,\mathrm{k}\Omega$
4. $R_1 = 10\,\mathrm{M}\Omega$, $R_2 = 10\,\mathrm{M}\Omega$
5. $R_1 = 1\,\mathrm{k}\Omega$, $R_2 = 220\,\Omega$
6. $R_1 = 220\,\Omega$, $R_2 = 1\,\mathrm{k}\Omega$

Record the measured values clearly in your notepad, and note any unexpected behaviour or measurement difficulties so you can compare them properly later.

(Task_I2_1_1)=
#### Task I2: Currents
Using DMM set in current mode, measure the currents in the circuits. You can try measuring in different places of the circuit.

Split the measurements evenly between both students, for example by letting one student measure one subset of resistor combinations and the other student measure the remaining subset, then compare the readings together.

Record the measured currents clearly in your notepad for the following resistor pairs:
1. $R_1 = R_2 = 100\,\Omega$
2. $R_1 = R_2 = 1\,\mathrm{k}\Omega$
3. $R_1 = R_2 = 100\,\mathrm{k}\Omega$
4. $R_1 = R_2 = 10\,\mathrm{M}\Omega$

Also note where you measured the current and what you observe, because careful records make comparison and troubleshooting much easier.

(Task_I3_1_1)=
#### Task I3: Preparation for tasks I4 and I5
Build following voltage dividers, keep them both on a breadboard and measure the voltages at the output of each divider.

Share this task equally: one student can build the left divider and the other student can build the right divider, and then you can swap roles for checking and measuring.

::::{div} hstack-grid

```{figure} images/voltage_divider_1k_220.svg
---
name: Voltage_divider_1k_220
height: 300px
---
Voltage divider circuit with 1kOhm and 220Ohm resistors
```

```{figure} images/voltage_divider_10k.svg
---
name: Voltage_divider_10k
height: 300px
---
Voltage divider circuit with 10kOhm resistors
```

::::

(Task_I4_1_1)=
#### Task I4: Connect divider outputs (parallel dividers)
Connect the two divider outputs and complete the circuit as shown below. Measure the output voltage and write down your result.

One student can make the circuit connection while the other verifies the wiring against the schematic and records the measured value; then switch roles in the next task.

```{figure} images/two_dividers_thevinen.svg
---
height: 300px
name: fig-two-dividers-thevinen-i4
---
Voltage dividers in parallel
```

(Task_I5_1_1)=
#### Task I5: Stack dividers in series
Connect the dividers in series as shown below. Measure the output voltage and write down your result.

Swap roles compared to Task I4 so the workload stays balanced between both students.

```{figure} images/stacked_dividers.svg
---
height: 300px
name: fig-stacked-dividers-i5
---
Voltage dividers in series
```

(Task_I6_1_1)=
#### Task I6: Measure $U_{rms}$ with a digital multimeter
Set up the function generator output so you can measure it with the DMM.

Keep the work balanced here as well: one student can set up the generator and connections, while the other student configures the DMM, records the readings, and checks whether the setup matches the instructions.

```{figure} images/function_generator_tb1_dmm.avif
---
name: fig-function-generator-tb1-dmm
class: no-invert
---
Function generator output connected to the DMM and to test board 1 
```

- Connect the function generator output to the DMM and to test board 1 using a BNC T-splitter and a BNC-to-banana adapter or cable.
- Make sure the BNC signal line goes to the red input and the shield to COM.
- Do not connect the BNC to any other components on the board.
- Set the function generator to a sine wave of $1\,\mathrm{kHz}$ and amplitude $U_{rms}=2\,\mathrm{V}$.
- Measure the voltage with the DMM.

**Questions to answer**:
1. Which type of voltage does the DMM display: $U_{pp}$, $U_p$, or $U_{rms}$?
2. If you set $U_{rms}=2\,\mathrm{V}$ on the generator, what does the DMM read?
3. Predict what happens if you connect an external load resistor: will the measured voltage increase, decrease, or stay the same?

(Task_I7_1_1)=
#### Task I7: Load effects and output setup
In Task A3 you discussed the function generator internal resistance and the effect of the load. Now verify this with measurements.

Divide the work evenly between the pair. For example, one student can focus on Part A and the other on Part B at first, and then both students should review and understand both sets of measurements before moving on.

```{figure} images/rotary_switch.svg
---
name: Rotary switch
height: 500px
---
Schematic of the rotary switch and BNC connectors on the test board.
```

##### Part A: Rotary switch load
- Keep the function generator at $U_{rms}=2\,\mathrm{V}$ and $1\,\mathrm{kHz}$ sine.
- Connect the rotary switch (SW1) as a load and measure $U_{rms}$ across it with the DMM.
- Find the switch setting (and resistance value) where the DMM reads the correct $U_{rms}$.
- Find the setting where the DMM reads approximately $\frac{U_{supply}}{2}$.
- Record the switch position and resistance value for both cases.

##### Part B: Output setup
The function generator has an OUTPUT SETUP that compensates for the expected load.
- Locate UTILITY / OUTPUT SETUP and choose either HIGH Z or $50\,\Omega$.
- With the test board connected and the rotary switch set to the position closest to $50\,\Omega$, decide which load setting gives the correct $U_{rms}$. Record the setting and the measured values.
- Disconnect the test board (no external load). Decide which load setting gives the correct $U_{rms}$. Record the setting and the measured values.

```{admonition} Hint
:class: hint, dropdown
HIGH Z assumes a very large load (e.g., DMM), while $50\,\Omega$ assumes a low load. Match the setting to the actual load so the displayed amplitude is correct.
```

### Compare and conclude
Use your notes, simulations, and measurements from the earlier parts of this manual. Draw the tables below in your notepad, fill them in using your earlier work, and use them to support your conclusions. These completed tables should be included in your onepager report for this lab. When possible, compare your conclusions with another pair at your table and discuss any differences before asking a TA.

#### Task C1: Compare predicted, simulated, and measured output voltages
```{admonition} Consider Tasks [A1](Task_A1_1_1), [S1](Task_S1_1_1), and [I1](Task_I1_1_1).
:class: note no-content
```

Draw the following table in your notepad and fill it in before answering the questions. Also add a short written note about what you notice when you compare the three columns.

| $R_1$    | $R_2$    | calculated $U_{out}$ [V] | simulated $U_{out}$ [V] | measured $U_{out}$ [V] |
|:---------|:---------|:-------------------------|:------------------------|:-----------------------|
| $100 Ω$  | $100 Ω$  |                          |                         |                        |
| $1 kΩ$   | $1 kΩ$   |                          |                         |                        |
| $100 kΩ$ | $100 kΩ$ |                          |                         |                        |
| $10 MΩ$  | $10 MΩ$  |                          |                         |                        |
| $1 kΩ$   | $220 Ω$  |                          |                         |                        |
| $220 Ω$  | $1 kΩ$   |                          |                         |                        |


1. For which resistor combinations do the calculated, simulated, and measured values of $U_{out}$ agree well?
2. For which combinations do you see the largest differences, and how large are those differences?
3. Which effects could explain the discrepancies: component tolerance, measurement accuracy, wiring, loading, or something else?
4. What general conclusion can you draw about when the simple voltage-divider formula is a good model?

#### Task C2: Relate voltage ratios to current levels
```{admonition} Consider Tasks [A1](Task_A1_1_1), [A2](Task_A2_1_1), [S1](Task_S1_1_1), [I1](Task_I1_1_1), and [I2](Task_I2_1_1).
:class: note no-content
```

Draw the following table in your notepad and fill it in before answering the questions. Use the completed table to support your conclusions, and include it in your onepager report.

| $R_1$    | $R_2$    | calculated $I_{out}$ [A] | simulated $I_{out}$ [A] | measured $I_{out}$ [A] |
|:---------|:---------|:--------------------------|:-------------------------|:------------------------|
| $100 Ω$  | $100 Ω$  |                           |                          |                         |
| $1 kΩ$   | $1 kΩ$   |                           |                          |                         |
| $100 kΩ$ | $100 kΩ$ |                           |                          |                         |
| $10 MΩ$  | $10 MΩ$  |                           |                          |                         |

1. Which resistor combinations give the same or nearly the same output voltage but very different currents?
2. Why can the voltage ratio stay the same while the current changes strongly?
3. Which divider values seem most practical for real measurements, and which values seem least practical?
4. What does this teach you about choosing resistor values, not only resistor ratios, in a real circuit?

#### Task C3: Explain loading and combined dividers
``` {admonition} Consider Tasks [A4](Task_A4_1_1), [S2](Task_S2_1_1), [I3](Task_I3_1_1), [I4](Task_I4_1_1), and [I5](Task_I5_1_1).
:class: note no-content
```

1. Why do the combined divider circuits not always behave like two isolated dividers?
2. How does connecting one divider to another create a loading effect?
3. How does the Thevenin equivalent help you predict what happens when the divider outputs are connected?
4. Which of the prepared dividers acts as the 'stronger' source, and how can you tell from the measurements and circuit values?

#### Task C4: Connect function-generator settings to divider behaviour
``` {admonition} Consider Tasks [A3](Task_A3_1_1), [I6](Task_I6_1_1), and [I7](Task_I7_1_1).
:class: note no-content
```

1. In what sense is the function generator together with its load another voltage-divider problem?
2. Why can the measured $U_{rms}$ change when the external load changes?
3. When should you choose HIGH Z output setup, and when should you choose $50\,\Omega$?
4. What practical rule will you use next time before trusting the amplitude shown on the function generator?

````{admonition} Deep dive: Proper resistance measurement
:class: deep-dive dropdown

```{admonition} Preparation
:class: tip
[Indirect resistance measurement](../../miscellaneous/appendix/resistance.ipynb#indirect-resistance-measurement)
```

##### Choosing the correct arrangement

You measured an unknown resistor with an ammeter of internal resistance $R_A = 2\,\Omega$ and a voltmeter of input resistance $R_V = 10\,\mathrm{M}\Omega$.

1. Compute the limiting resistance

   $$ R_{gr} = \sqrt{R_A R_V} $$

   and state, for each of the resistor values you used in Task I1 ($100\,\Omega$, $1\,\mathrm{k}\Omega$, $100\,\mathrm{k}\Omega$, $10\,\mathrm{M}\Omega$), which of the two arrangements (current-correct or voltage-correct) you should pick.

2. For a nominal $R_X = 100\,\Omega$, predict the measured resistance $R_{Xp}$ and the relative method error $\delta$ when:
    - the current-correct circuit is used (the wrong choice for this $R_X$),
    - the voltage-correct circuit is used (the right choice for this $R_X$).

3. Repeat point 2 for $R_X = 100\,\mathrm{k}\Omega$, swapping the roles (current-correct is now the right choice and voltage-correct the wrong one). Comment on which mistake (wrong arrangement for a small $R_X$ vs. wrong arrangement for a large $R_X$) leads to a bigger relative error in *your* setup, and explain in one or two sentences why this depends on the ratio between $R_X$, $R_A$, and $R_V$.


```{admonition} Check your answer
:class: answer, dropdown
For $R_A = 2\,\Omega$ and $R_V = 10\,\mathrm{M}\Omega$, $R_{gr} \approx 4.5\,\mathrm{k}\Omega$. So $100\,\Omega$ and $1\,\mathrm{k}\Omega$ go to voltage-correct, while $100\,\mathrm{k}\Omega$ and $10\,\mathrm{M}\Omega$ go to current-correct.

For $R_X = 100\,\Omega$, current-correct (wrong) gives $R_{Xp} = R_X + R_A = 102\,\Omega$, $\delta \approx +2\%$. Voltage-correct (right) gives $\delta \approx -10^{-6}$, essentially perfect.

For $R_X = 100\,\mathrm{k}\Omega$, voltage-correct (wrong) gives $R_{Xp} = R_X \parallel R_V \approx 99.01\,\mathrm{k}\Omega$, $\delta \approx -1\%$. Current-correct (right) gives $\delta \approx +2\cdot10^{-5}$, also essentially perfect.

So with $R_X$ on the same side of $R_{gr}$ in both cases, the errors of the *wrong* choice end up of comparable magnitude (roughly 1 to 2%). The picture would be much harsher only if $R_X$ approached $R_V$ (for example $10\,\mathrm{M}\Omega$), where the wrong, voltage-correct arrangement would short a sizeable fraction of the current through the voltmeter.
```

---

##### Measure the same resistor two ways

Use **two** resistors from your kit: a "small" one $R_X = 100\,\Omega$ and a "large" one $R_X = 100\,\mathrm{k}\Omega$. These two values sit on opposite sides of the limiting resistance $R_{gr}$, so they should pick out opposite arrangements.

For each resistor:

1. **Direct (reference) measurement.** Measure $R_X$ in 2-wire ohmmeter mode and record it as $R_{X,\Omega}$.

2. **Current-correct circuit.** Build the circuit shown as *Properly measured current* in the appendix, using a stable DC source (e.g., $5\,\mathrm{V}$). Record $U_V$, $I_A$, and compute $R_{Xp}^{(I)} = U_V/I_A$.

3. **Voltage-correct circuit (voltmeter directly across $R_X$).** Only move the voltmeter. Record $U_V$, $I_A$, and compute $R_{Xp}^{(V)} = U_V/I_A$.

4. Fill in the following table in your notepad:

   | $R_X$ nominal | $R_{X,\Omega}$ (DMM ohms) | $R_{Xp}^{(I)}$ (current-correct) | $R_{Xp}^{(V)}$ (voltage-correct) | which one matches $R_{X,\Omega}$ better? |
   |:---:|:---:|:---:|:---:|:---:|
   | $100\,\Omega$    |   |   |   |   |
   | $100\,\mathrm{k}\Omega$ |   |   |   |   |

5. **Cross-check with theprediction.** For each row, compute the expected sign of the error and check that the column you marked as "matches better" is consistent with that sign.

6. **Reflect (one short paragraph in your notepad).**
    - Did the 2-wire ohmmeter agree with the better of the two voltmeter and ammeter results? If not, where do you think the residual discrepancy comes from (lead resistance, contact resistance, source ripple, DMM range)?
    - If you were asked to measure a $50\,\mathrm{m}\Omega$ shunt or a $50\,\mathrm{M}\Omega$ insulation resistor with this same bench, which arrangement would you pick?

```{Tip} 
At $5\,\mathrm{V}$, the current through the $100\,\Omega$ resistor is $50\,\mathrm{mA}$, which the DMM and the resistor can take briefly. Keep the source connected only as long as you need to read the meters. Through the $100\,\mathrm{k}\Omega$ resistor the current is only $50\,\mu\mathrm{A}$, well within any DMM current range.
```

````

````{admonition} Deep dive: Y-Δ transform on a real bridge
:class: deep-dive dropdown

```{admonition} Preparation
:class: tip
[Y-Δ transform](../../miscellaneous/appendix/resistance.ipynb#simplification-of-complex-resistor-circuits-y-transform)
```

The Y-Δ transform is a handy shortcut for circuits that **cannot** be reduced with series and parallel alone. The five-resistor bridge below is the smallest example: a single Δ→Y substitution turns it into a network you already know how to solve by inspection.

##### Reduce an (un)balanced bridge

Consider the following five-resistor unbalanced bridge driven from a $5\,\mathrm{V}$ source between input nodes $X$ and $Z$:

- $R_1 = 1\,\mathrm{k}\Omega$ from $X$ to node $A$
- $R_2 = 2.2\,\mathrm{k}\Omega$ from $X$ to node $B$
- $R_3 = 1\,\mathrm{k}\Omega$ between $A$ and $B$ (the "bridge" resistor)
- $R_4 = 1\,\mathrm{k}\Omega$ from $A$ to $Z$
- $R_5 = 2.2\,\mathrm{k}\Omega$ from $B$ to $Z$

The output is the voltage between $A$ and $B$.

1. Convince yourself that this network **cannot** be reduced with only series and parallel rules: identify the two "innermost" pairs of resistors and explain in one sentence why neither pair is in pure series or pure parallel.

2. Pick the Δ formed by $R_1$, $R_2$, $R_3$. Apply the **delta-to-star** transformation to find the equivalent star $R_X^\star$, $R_A^\star$, $R_B^\star$ that connects $X$, $A$, $B$ to a new internal node $N$.

3. Redraw the bridge with the Y you just computed in place of the original Δ. The redrawn circuit should now be a series-parallel network. Compute:
    - the equivalent resistance seen by the source between $X$ and $Z$,
    - the voltage at the new node $N$,
    - the voltages at $A$ and $B$,
    - $U_{out} = U_A - U_B$.

4. **Swap $R_3$.** What would happen to $U_{out}$ if you replaced $R_3$ with a different value (e.g., $2.2\,\mathrm{k}\Omega$, or even a wire)? Argue without redoing the Y-Δ calculation.

5. **Swap $R_5$.** What would happen to $U_{out}$ if you replaced $R_5$ with $1\,\mathrm{k}\Omega$? Predict qualitatively (stays zero, grows, changes sign?) before crunching numbers.


```{admonition} Check your answer
:class: answer, dropdown
- $R_{eq} \approx 1.375\,\mathrm{k}\Omega$
- $U_N \approx 3.10\,\mathrm{V}$
- $U_A \approx 2.50\,\mathrm{V}$
- $U_B \approx 2.50\,\mathrm{V}$
- $U_{out} \approx 0\,\mathrm{V}$ (the bridge is balanced for these values: $R_1/R_4 = R_2/R_5 = 1$)
- Swap $R_3$: $U_{out}$ does **not** change. At balance no current flows through the bridge resistor, so its value is irrelevant.
- Swap $R_5 \to 1\,\mathrm{k}\Omega$: balance breaks ($R_2/R_5 = 2.2 \ne R_1/R_4 = 1$); $U_B$ drops below $2.50\,\mathrm{V}$ and $U_{out}$ becomes positive, about $0.43\,\mathrm{V}$.
```

---

##### Measure the bridge

Reuse the breadboard, DMM, and DC source from Tasks I1 to I5. The whole bridge fits on a single half of the breadboard.

1. **Build the bridge** with the five resistors above. Before powering it, label nodes $X$, $Z$, $A$, $B$ in your notepad sketch. Having a name for every node will save time later.

2. **Measure the unloaded source.** Set the supply to $5\,\mathrm{V}$ and confirm with the DMM. Record the actual value you will use in the comparison; it is rarely exactly $5.000\,\mathrm{V}$.

3. **Measure each output voltage individually:**
    - $U_A$ from $A$ to $Z$ (ground reference),
    - $U_B$ from $B$ to $Z$,
    - $U_{out} = U_A - U_B$, *and* also measure $U_{out}$ directly between $A$ and $B$ as a cross-check that the DMM does not lie about small differences of large numbers.

4. **Measure the equivalent input resistance** $R_{in}$ from $X$ to $Z$ with the supply *disconnected*, in 2-wire ohmmeter mode. Compare to the value you predicted with the Y-Δ method.

5. Fill in the following comparison table:

   | quantity | predicted (Y-Δ) | measured |
   |---|---|---|
   | $U_A$ |   |   |
   | $U_B$ |   |   |
   | $U_{out} = U_A - U_B$ |   |   |
   | $R_{in}$ from $X$ to $Z$ |   |   |

6. Do the measured $U_A$, $U_B$, $U_{out}$, and $R_{in}$ agree with the Y-Δ prediction? If not, is the residual gap explained by ordinary tolerances, or is something else going on?


````
