#### Task A5: Distinguish $U_{pp}$, $U_p$, $U_{rms}$
While generating a sine wave, the function generator is most often set to $U_{pp}$ (volt peak-to-peak). The DMM measures $U_{rms}$ (volt root-mean-square).  

```{figure} images/Vpp_Vp_Vrms.png
---
name: vpp-vrms
class: no-invert
---
Oscilloscope showing the sine wave
```
The function generator generates the signal above. Derive the amplitude of the signal in $U_{rms}$, $U_p$, and $U_{pp}$. Also derive the frequency of the sine wave.

```{admonition} Check your answer
:class: answer, dropdown
From a sine wave with amplitude $U_p = 1\,\mathrm{V}$, the peak-to-peak value is $U_{pp} = 2\,\mathrm{V}$. The $U_{rms}$ is the continuous DC signal having the same power as the time-varying AC signal, the quadratic mean. The $U_{rms}$ of this sine wave is $U_{rms} = \frac{1}{\sqrt{2}}\,\mathrm{V} \approx 0.71\,\mathrm{V}$.  
```


#### Task A6: Which DMM range gives the best accuracy?

A $5\,\frac{1}{2}$-digit DMM shows $5$ full digits plus one extra leading digit that can only be $0$ or $1$.
So on the highest range it counts from $0$ to $199999$.
The accuracy of the $5\,\frac{1}{2}$-digit DMM is specified by the manufacturer as $\pm(0.025\% + 0.005\%) = \pm(\text{reading}\% + \text{range}\%)$.

1. Calculate the accuracy for a measurement of $1.2\,\mathrm{V}$ on the $2\,\mathrm{V}$ range.
2. Calculate the accuracy for a measurement of $1.2\,\mathrm{V}$ on the $20\,\mathrm{V}$ range.
3. Explain which range is preferred for optimally measuring $1.2\,\mathrm{V}$ and why.

For this task, use the following details about the ranges:
- $2\,\mathrm{V}$ range ($3\,\frac{1}{2}$-digit DMM): max $1.999\,\mathrm{V}$, min $0.001\,\mathrm{V}$.
- $20\,\mathrm{V}$ range ($3\,\frac{1}{2}$-digit DMM): max $19.99\,\mathrm{V}$, min $0.01\,\mathrm{V}$.
- With the same number of digits, each range has a different number of decimal places.

```{admonition} Check your answer
:class: answer, dropdown
For $1.2\,\mathrm{V}$, the accuracy is $\pm 0.025\% \cdot 1.2 + 0.005\% \cdot 2\,\mathrm{V} = \pm0.4\,\mathrm{mV}$.
```
