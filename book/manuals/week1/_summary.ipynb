# Week 1


In this week, you will explore how basic circuits and measurement instruments behave in practice.

- analyze voltage dividers, current levels, and loading effects in simple and connected circuits
- use the multimeter, function generator, and oscilloscope to measure and interpret electrical signals
- compare calculated, simulated, and measured results to understand differences between theory and practice
- investigate AC and DC coupling, triggering, and the behaviour of RC filters in time and frequency domains
