#### Introduction to the oscilloscope

Last week, our focus was on the Digital Multimeter (DMM). The DMM is a helpful tool that can tell us the voltage, resistance and current of a circuit (plus additional features, which you can read about in the [DMM intro page](https://nb-tudelft.github.io/Electronic_Instrumentation/EI_manuals/intro_DMM_Siglent%2Bhandheld.html)). 

Even though the DMM is a very useful tool for electronics, it only gives you a single value for a reading. This means that if you are trying to measure a signal that is changing over time, the DMM won't be able to visualise it. As you saw last week, when you measure an alternating current (AC) signal, such as a sine wave, using the DMM, the reading will only show the Urms of the signal. 

For AC signals, it may be more interesting to see how the voltage changes over time, rather than just the Urms. This would give us more insight into the shape and frequency of the signal. This is exactly what the **oscilloscope (often called scope for short)** allows us to do!

Similar to how we can use resistances to derive the transfer function of a voltage divider, we can use impedances to derive the transfer function of a filter. Lets look at the example in the figure below. 

```{figure} images/High_pass_RC_filter.jpg
---
height: 200px
name: High pass RC filter
---
High pass RC filter
```

Lets refer to the impedance of the capacitor as $Z_1$ and the impedance of the resistor as $Z_2$. Just like for a simple voltage divider with resistors, the output voltage will be as follows: $U_{out} = \frac{Z_2}{Z_1+Z_2}\times U_{in}$. Since we know that $Z_1=\frac{1}{jωC}$ and $Z_2=R$, we can plug in these values to obtain the expression: $U_{out} = \frac{R}{\frac{1}{jωC}+R}\times U_{in}$. Dividing the numerator and denominator by $R$ results in the final expression:

$$U_{out} = \frac{1}{\frac{1}{jωRC}+1}\times U_{in}$$

To find out whether this filter is a low pass or high pass filter, we can look at the behaviour of the circuit for very low and very high frequencies. 

As the frequency of the input signal becomes very small (i.e. as $ω \rightarrow 0$), we will have $U_{out} = \frac{1}{\frac{1}{0}+1}\times U_{in} = \frac{1}{\infty+1}\times U_{in}$, which simplifies to $U_{out}=0$. Hence, this filter blocks low frequencies. 

As the frequency of the input signal becomes very large (i.e. as $ω \rightarrow \infty$), we will have $U_{out} = \frac{1}{\frac{1}{\infty}+1}\times U_{in} = \frac{1}{0+1}\times U_{in}$, which simplifies to $U_{out}= U_{in}$. This means that this filter will pass high frequencies. 

As the filter allows high frequencies to pass and blocks low frequencies, we would refer to it as a **high pass filter**. 

To visualise the behaviour of a filter, it is helpful to look at a graph showing the frequency response of the filter. Typically, the x-axis will show frequency in Hz with a logarithmic scale and the y-axis will show the gain of the filter in decibels. The formula for gain in decibels is as follows:

$$ Gain(dB) = 20\log|\frac{U_{out}}{U_{in}}| $$

The frequency response of a high pass filter can be visualised in the figure below. As we can see in the figure, the filter blocks low frequencies and lets through high frequencies. The gain of the signal (the amount of signal let through the filter), increases as the frequency increases. 
```{figure} images/Freq_response_HPF.svg
---
height: 300px
name: High pass RC filter
---
Frequency response of a high pass filter
```
An important point on the graph of the frequency response is the **cutoff frequency ($f_c$)** of the filter. This is also referred to the **-3dB** point of the filter, as it is the frequency at which the gain in decibels is -3dB. The cutoff frequency of a high pass filter acts as a threshold, indicating the frequency above which the majority of the signal will be passed. At the -3dB point, $|\frac{U_{out}}{U_{in}}| \approx \frac{1}{\sqrt{2}}$. We can find the cutoff frequency ($f_c$) either by looking at the frequency response of the filter and reading off the frequency corresponding to a gain of -3dB, or by using the transfer function as follows:

$|\frac{U_{out}}{U_{in}}| \approx \frac{1}{\sqrt{2}}$

$\frac{1}{\sqrt{(ωRC)^2+1}} =\frac{1}{\sqrt{2}}$

$\sqrt{ω^2R^2C^2+1} = \sqrt{2}$

$ω^2R^2C^2 = 1$

$ωRC = 1$

$ω_c= \frac{1}{RC}$    
$f_c = \frac{1}{2\pi RC}$

## Anticipate 1: Effect of AC/DC coupling and signal frequency on the measured signal

When you input a signal into the oscilloscope, there are two measurement options:
- AC coupling 
- DC coupling 

When you measure in DC coupling mode, the oscillocope shows the whole signal. In contrast, when you measure in AC coupling mode, the oscilloscope will take away the average value of the signal. This means that it only shows you the part of the signal that is changing. 

How does the oscilloscope do this? As seen in the figure below, when you choose to measure in DC coupling mode, the input signal goes directly to the oscilloscope. This means that the circuit connecting the input signal to the oscillscope behaves similarly to a simple circuit with a resistor. Hence, the circuit will let through all frequencies/the whole signal. 

In AC coupling mode, the circuit connecting the input signal to the oscilloscope contains an extra capacitor. This means the circuit behaves similarly to a high-pass RC filter and thus, will block low frequency components in the signal. By definition, DC components are not changing, and hence have a frequency of 0Hz. Accordingly, in AC coupling mode, the DC component of the signal will be blocked. 

```{figure} images/AC_DC_coupling.png
---
height: 300px
name: AC/DC coupling schematic 
---
Schematic of AC vs DC coupling modes on the oscilloscope
```

The effect of AC vs DC coupling modes can be visualised well by connecting the function generator to the oscilloscope and inputting block waves with various duty cycles. 

Duty cycle refers to the percentage of time a block wave is in the high state. All the block waves you have seen before have likely had a 50% duty cycle. This means that the block wave is in the high state half of the time and in the low state for the other half of the time. Since a 50% duty cycle wave is symmetrical, the average or middle value of the signal will be exactly halfway between the high and low values of the signal. As seen in the figure below, when a block wave has a 75% duty cycle, it is in the high state 75% of the time. Similarly, a 25% duty cycle block wave is in the high state 25% of the time. 

```{figure} images/Duty_cycle.png
---
height: 300px
name: Block waves with different duty cycles
---
Block waves with various duty cycles
```

## Anticipate 2: Triggering 

The oscilloscope is constantly acquiring measurements of the input signal, so how does it know at what point it should start showing the signal on the display? In other words, how does it know when to trigger? This is determined by the trigger settings. The trigger settings we will focus on are: 
- Source: here you can select a channel for the oscilloscope to "look at" when triggering
- Slope: this is the slope that the input signal must have for the oscilloscope to "start drawing"/trigger  
- Trigger level: this is the threshold that the input signal must reach for the oscilloscope to trigger. If the trigger level is never reached, then the oscilloscope can't trigger. This results in the signal jumping around along the time axis 

For example, if a 4 Upp sine wave is connected to channel 1 of the oscilloscope and you select the trigger source as channel 1, slope as rising and the trigger level as 1 V, you would expect to see the following signal.

The scope probe is a type of cable used to measure signals. One side of the scope probe can be connected to the oscilloscope in the same manner as a BNC/coax cable. While the scope probe and coax cable share many similarities, the coax cable cable has a slight low-pass filtering effect. This represents a disadvantage when measuring high frequency signals, as the coax cable will attenuate (reduce the intensity) of the signal it is measuring. In contrast, the scope probe contains an additional capacitor to offset this low pass filtering effect. As this low pass filtering effect is quite slight, if the circuit you are measuring is already a low pass filter, then there will be no noticeable difference between using a coax cable and the scope probe. 

Today there will be no noticeable difference when measuring with a coax cable and a scope probe. We are simply choosing to use the scope probe to offer you an opportunity to gain familiarity with measuring with this tool.
