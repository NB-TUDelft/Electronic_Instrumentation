#### Introduction to the oscilloscope

Last week, our focus was on the Digital Multimeter (DMM). The DMM is a helpful tool that can tell us the voltage, resistance and current of a circuit (plus additional features, which you can read about in the [DMM intro page](https://nb-tudelft.github.io/Electronic_Instrumentation/EI_manuals/intro_DMM_Siglent%2Bhandheld.html)). 

Even though the DMM is a very useful tool for electronics, it only gives you a single value for a reading. This means that if you are trying to measure a signal that is changing over time, the DMM won't be able to visualise it. As you saw last week, when you measure an alternating current (AC) signal, such as a sine wave, using the DMM, the reading will only show the Vrms of the signal. 

For AC signals, it may be more interesting to see how the voltage changes over time, rather than just the Vrms. This would give us more insight into the shape and frequency of the signal. This is exactly what the **oscilloscope (often called scope for short)** allows us to do!


- The *input channels* are where we connect the signal we want to view on the oscilloscope
- We can visualise the signal on the *display* region 
- *Horizontal configuration* 
    - The top knob allows us to control the size of the time axis that we see on the display. Twisting clockwise "zooms in" in time
    - The bottom knob allows us to move our signal to the left or right
- *Vertical configuration* 
    - The top knob controls the size of the voltage axis that we see on the display. Twisting clockwise "zooms in" in voltage 
    - The bottom knob allows us to move our signal up or down
- The oscilloscope is constantly recording the input signal, but it needs to know at what point it should start "drawing" the signal on the display. This is controlled by the *trigger settings*

Today we will dive deeper into these basic settings and gain familiarity with using the scope.

____

#### RC filters


## Anticipate 1: Effect of AC/DC coupling and signal frequency on the measured signal

When you input a signal into the oscilloscope, there are two measurement options:
- AC coupling 
- DC coupling 

When you measure in DC coupling mode, the oscillocope shows the whole signal. In contrast, when you measure in AC coupling mode, the oscilloscope will take away the average value of the signal. This means that it only shows you the part of the signal that is changing. 

How does the oscilloscope do this? As seen in the figure below, when you choose to measure in DC coupling mode, the input signal goes directly to the oscilloscope. This means that the circuit connecting the input signal to the oscillscope behaves similarly to a simple circuit with a resistor. Hence, the circuit will let through all frequencies/the whole signal. 

In AC coupling mode, the circuit connecting the input signal to the oscilloscope contains an extra capacitor. This means the circuit behaves similarly to a high-pass RC filter and thus, will block low frequency components in the signal. By definition, DC components are not changing, and hence have a frequency of 0Hz. Accordingly, in AC coupling mode, the DC component of the signal will be blocked. 

```{figure} images/AC_DC_coupling.png
---
width: 400px
name: AC/DC coupling schematic 
---
Schematic of AC vs DC coupling modes on the oscilloscope
```

The effect of AC vs DC coupling modes can be visualised well by connecting the function generator to the oscilloscope and inputting block waves with various duty cycles. 

Duty cycle refers to the percentage of time a block wave is in the high state. All the block waves you have seen before have likely had a 50% duty cycle. This means that the block wave is in the high state half of the time and in the low state for the other half of the time. Since a 50% duty cycle wave is symmetrical, the average or middle value of the signal will be exactly halfway between the high and low values of the signal. As seen in the figure below, when a block wave has a 75% duty cycle, it is in the high state 75% of the time. Similarly, a 25% duty cycle block wave is in the high state 25% of the time.
