#### Task B2: AC/DC coupling modes on the oscilloscope
When you input a signal into the oscilloscope, there are two measurement options:
- AC coupling 
- DC coupling 

When you measure in DC coupling mode, the oscillocope shows the whole signal. In contrast, when you measure in AC coupling mode, the oscilloscope will take away the average value of the signal. This means that it only shows you the part of the signal that is changing. 

How does the oscilloscope do this? As seen in the figure below, when you choose to measure in DC coupling mode, the input signal goes directly to the oscilloscope. This means that the circuit connecting the input signal to the oscillscope behaves similarly to a simple circuit with a resistor. Hence, the circuit will let through all frequencies/the whole signal (until about 20 MHz, the bandwidth of the oscilloscope). 

In AC coupling mode, the circuit connecting the input signal to the oscilloscope contains an extra capacitor. This means the circuit behaves similarly to a high-pass RC filter and thus, will block low frequency components in the signal. By definition, DC components are not changing, and hence have a frequency of 0 Hz. Accordingly, in AC coupling mode, the DC component/average value of the signal will be blocked.

The effect of AC vs DC coupling modes can be visualised well by connecting the function generator to the oscilloscope and inputting block waves with various duty cycles. 

Duty cycle refers to the percentage of time a block wave is in the high state. All the block waves you have seen before have likely had a 50% duty cycle. This means that the block wave is in the high state half of the time and in the low state for the other half of the time. Since a 50% duty cycle wave is symmetrical, the average or middle value of the signal will be exactly halfway between the high and low values of the signal. As seen in the figure below, when a block wave has a 75% duty cycle, it is in the high state 75% of the time. Similarly, a 25% duty cycle block wave is in the high state 25% of the time.
