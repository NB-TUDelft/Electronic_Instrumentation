
An opamp has a large gain for the difference between its inputs. In a linear amplifier, negative feedback keeps that difference small. In a **comparator**, the output moves towards one of two limiting levels:

* If $U_+ > U_-$, the output moves towards its high level, $U_H$.
* If $U_+ < U_-$, the output moves towards its low level, $U_L$.

The comparator tells you which input is higher. Its output levels depend on the chip, supply, and load. They need not equal the supply voltages.

Here the signal $U_{in}$ is connected to the positive input, and the negative input is grounded.

```{admonition} Why is $R_a$ included? Would it work without it?
We include $R_a$ here because you will need it when you add [hysteresis](comparator-hysteresis-6-1) to the circuit. Together with $R_b$, it will determine the switching thresholds.

A series resistor can also limit current through input protection structures if they conduct. Whether this matters depends on the opamp you use.
```

##### Noise near the switching threshold

Real signals contain small, unwanted voltage fluctuations called **noise**. When a slowly changing input is close to the comparator's switching threshold, noise can move it above and below that threshold several times. The output then switches high and low repeatedly instead of making one transition. This behaviour is often called **chatter**.

```{figure} images/oscillator-noise.svg
---
name: fig-oscillator-noise
height: 380px
---
The same noisy input produces extra switches without hysteresis and a single switch with hysteresis in this illustrative example.
```

If you use the output to count events or measure frequency, these extra switches can give an incorrect result.

A comparator with **hysteresis** uses two switching thresholds. After the output switches high, the input must reach the lower threshold before the output can switch low again. Small fluctuations around the upper threshold therefore do not cause repeated switches.

##### Hysteresis
The response in {numref}`fig-oscillator-noise` depends on which threshold the input crossed last. A comparator with this behaviour is called a **Schmitt trigger**. Follow the two paths in {numref}`fig-oscillator-hysteresis` to see how the output depends on both the input voltage and its previous state.

```{figure} images/oscillator-hysteresis.svg
---
name: fig-oscillator-hysteresis
height: 320px
---
The comparator switches at different input voltages depending on whether the input is increasing or decreasing.
```

The horizontal axis shows the input voltage $U_{in}$. The vertical axis shows the comparator output $U_q$.

* **Increasing input, orange:** starting with a low output, follow the lower line to the right. The output switches high at the upper threshold, $U_{T,up}$.
* **Decreasing input, blue:** starting with a high output, follow the upper line to the left. The output switches low at the lower threshold, $U_{T,down}$.

For example, at $U_{in}=0$, the output can be high or low, depending on which threshold was crossed last.

##### Add positive feedback

To produce this behaviour, connect $R_b$ from the output to the positive input, as shown in the right-hand circuit of {numref}`fig-oscillator-comparator`. This **positive feedback** reinforces the current output state.

```{figure} images/oscillator-comparator.svg
---
name: fig-oscillator-comparator
height: 320px
---
OA2 without feedback (left) and with positive feedback through $R_b$ (right). The input signal enters through $R_a$.
```

The voltage at the positive input is a weighted combination of $U_{in}$ and $U_q$, the comparator output. Neglecting input current, Kirchhoff's current law gives

$$
\frac{U_{in}-U_+}{R_a}+\frac{U_q-U_+}{R_b}=0.
$$

At the ideal switching boundary, $U_+=U_-=0$. Use this boundary separately for $U_q=U_H$ and $U_q=U_L$ in [Task A2](Task_A2_6_1).
