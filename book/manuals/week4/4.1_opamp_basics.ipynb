#### Task B1: Operational amplifier basics
This manual builds three circuits around a single eight-pin opamp chip (TL072CP or LM358P). The unifying idea is the **virtual short**: in a closed feedback loop, the opamp drives its output until both inputs sit at the same voltage. From that single fact you can derive the gain of every circuit you will build today.

##### The chip and its supply

The chip has two opamps inside one DIP-8 package. During this practical you will use one of them. 

```{figure} images/opamp_DIP8.svg
---
name: fig-opamp-pinout
height: 400px
---
DIP-8 packaged opamp pinout
```


The relevant pins are:

* pin 1: output of opamp 1 (the one you will use).
* pin 2: inverting input of opamp 1.
* pin 3: non-inverting input of opamp 1.
* pin 4: negative supply, $V_- = -12\,\mathrm{V}$.
* pin 8: positive supply, $V_+ = +12\,\mathrm{V}$.

The two power pins may not be visible in opamp schematics below, but you must wire them every time. **The opamp will not work without both rails connected, and it can be damaged if the polarity is reversed.**

### Anticipate

You will build three opamp configurations today: non-inverting, inverting, and current-to-voltage (transimpedance). The five tasks below let you predict every measurement you are about to make. Keep your derivations and numbers in your notepad: in the Compare and Conclude section at the end of the manual you will compare them against simulation and against the bench measurement.


(Task_A1_4_1)=
#### Task A1: Derive non-inverting gain from the virtual short
```{admonition} Estimated time: 15 min
:class: estimated-time no-content
```

Replace the input voltage source with a current source. Set $R_f = 100\,\mathrm{k}\Omega$. Run a DC sweep of $I_{in}$ from $0$ to $2\,\mathrm{m A}$.

1. Plot $U_{out}$ vs. $I_{in}$. Find the value of $I_{in}$ where the output voltage stops following the ideal $U_{out} = -I_{in} R_f$ line. Record it.
2. Add a probe on the inverting input. Plot $U_{-}$ on the same axes. As long as the circuit is in the linear region, $U_{-}$ should sit at zero (the virtual ground holds). Where does $U_{-}$ start to drift away from zero?

The point where $U_{-}$ stops sitting at zero is the moment the **golden rules break**. The output cannot reach the voltage the formula demands, so the feedback loop loses its grip and the inverting input is no longer pinned to ground. This is the single cleanest demonstration of when the ideal opamp model fails. Keep this picture in your head, you will see it again in Manual 4.2.

This task is the textbook demonstration of why a buffer is a useful circuit, even though its gain is just 1.

1. Convert your non-inverting amplifier into a buffer: short $R_a$ (replace it with a wire) and remove $R_b$ (leave its pad open). Confirm the gain is now exactly 1 by repeating the measurement from [Task I1](Task_I1_4_1).

You should see two very different scope traces in steps 2 and 3, even though the source signal and the series resistor are identical.

**Questions to answer in your notepad:**

* Why is the amplitude in step 3 much smaller than the function-generator setting?
* Why does the buffer in step 2 restore (almost) the full amplitude?
* Where would you use a buffer in a real measurement system?

```{admonition} Hint
:class: hint, dropdown
The scope input has a finite input impedance (about $1\,\mathrm{M}\Omega$). With a $1\,\mathrm{M}\Omega$ source resistor in series, you have built a 1:1 voltage divider and lost half the signal. The buffer presents an essentially infinite input impedance to the source, and a near-zero output impedance to the scope, so almost no signal is lost.
```

(Task_I4_4_1)=
#### Task I4: Build and measure the inverting amplifier
```{admonition} Estimated time: 25 min
:class: estimated-time no-content
```
