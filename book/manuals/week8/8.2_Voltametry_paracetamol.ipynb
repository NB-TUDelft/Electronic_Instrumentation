1. **Phosphate Buffered Saline (PBS) - ph = 7.4**
   > PBS is widely used in biological and biochemical experiments to maintain a stable pH and osmotic balance, replicating the conditions in living organisms.
1. **Paracetamol - 0.5mg/ml in PBS**
    > Samples with Paracetamol were prepared using a generic 500mg Paracetamol tablet purchased in Kruidvat. The concentration of 0.5 mg/ml in PBS was chosen for its optimal performance across a wide range of experimental parameters.

During this assignement you will do most experiments on these two samples. The optimal set of experimental parameters, $R_f$, <font color="#a25809">**DAC B**</font>, $U_{cell,min}$ and $U_{cell,max}$ are not necessarily the same. You will have to find the best combination by perfroming the sweep multiple times while adapting these parameters.

* This is your last warning, make sure your code can easily change these parameters in order to quickly find the optimal set.
* Also try to build in an option to perform cyclic voltammetry, i.e. letting the sweep run for multiple cycles.

```{attention}
Always wire the $\pm 12\,\mathrm{V}$ supply with the supply **off**. Verify the polarity at pin 8 (positive) and pin 4 (negative) with the DMM before connecting the chip. Reversing the rails will destroy the opamp within seconds.
```


(Task_I1_8_2)=
#### Task I1: Connect the measurement cell
```{admonition} Estimated time: 15 min
:class: estimated-time no-content
```

The measurement cell you are going to use during this assingment consists of two parts. The solid board can be direclty plugged in a breadboard. You could use a separate board or th main board on the ALPACA. Three pins (CE, RE, Ref) connect to the three legs on the disposable adapter. You are supposed to put liquid on the round part, making sure that all three legs are covered. Use a different adapter for every solution.
