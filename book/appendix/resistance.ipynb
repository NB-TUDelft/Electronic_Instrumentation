# Resistance recap


In electronics, resistance is the opposition to the flow of electric current in an electrical circuit. Resistance is measured in ohms ($\Omega$). There is also a quantity that measures how easily current can flow through a material, called conductance $G$, with units of siemens ($\mathrm{S}$). In basic electronics and electrotechnics, there are 3 main laws that are essential for analyzing, describing, and calculating parameters of electronic circuits. These are Ohm's law and Kirchhoff's laws, 1st (junction rule) and 2nd (loop rule). It is important to keep the idea of these laws in mind as a concept, not a bunch of formulas. Someone smart used to say these are enough to get a PhD in electrical engineering.

Ohm's law states that the voltage $U$ across a resistor is proportional to the current $I$ through it: $U = RI$. This means that for a fixed resistance $R$, a larger voltage $U$ causes a larger current $I$. Kirchhoff's current law (junction rule) states that the sum of currents entering a node must equal the sum of currents leaving it. Kirchhoff's voltage law (loop rule) states that the sum of all voltage rises and drops around a closed loop must be zero. These laws express conservation of charge and conservation of energy in electrical circuits.

All electronic circuits and further formulas can be derived from these laws. Let us first look at the parallel case, then the series case, and finally use that result to derive the voltage divider formula.

For two resistors in parallel, the voltage across both resistors is the same:
$$
U = U_1 = U_2
$$
Using Ohm's law, the current in each branch is:
$$
I_1 = \frac{U}{R_1}, \qquad I_2 = \frac{U}{R_2}
$$
Kirchhoff's current law tells us that the total current is the sum of the branch currents:
$$
I = I_1 + I_2
$$
Substituting the expressions for $I_1$ and $I_2$ gives:
$$
I = \frac{U}{R_1} + \frac{U}{R_2}
$$
$$
I = U\left(\frac{1}{R_1} + \frac{1}{R_2}\right)
$$
For the equivalent resistance $R_{eq}$ of the same network, we also have the following:
$$
I = \frac{U}{R_{eq}}
$$
So:
$$
\frac{U}{R_{eq}} = U\left(\frac{1}{R_1} + \frac{1}{R_2}\right)
$$
Dividing both sides by $U$ gives:
$$
\frac{1}{R_{eq}} = \frac{1}{R_1} + \frac{1}{R_2}
$$
For two resistors, this can be rewritten as:
$$
R_{eq} = \frac{R_1R_2}{R_1 + R_2}
$$
Now compare this with the series case. For two resistors in series, the same current flows through both resistors:
$$
I = I_1 = I_2
$$
Kirchhoff's voltage law tells us that the total voltage is the sum of the voltage drops:
$$
U = U_1 + U_2
$$
Using Ohm's law for each resistor gives:
$$
U = IR_1 + IR_2
$$
$$
U = I(R_1 + R_2)
$$
For the equivalent resistance $R_{eq}$ we also write:
$$
U = IR_{eq}
$$
Therefore:
$$
R_{eq} = R_1 + R_2
$$
This series result leads smoothly to the voltage divider formula. If $R_1$ and $R_2$ are connected in series across an input voltage $U_{in}$ and the output is measured across $R_2$, then the same current flows through both resistors:

$$
I = \frac{U_{in}}{R_1 + R_2}
$$
The output voltage is the voltage across $R_2$:
$$
U_{out} = IR_2
$$
Substituting the expression for $I$ gives the following:
$$
U_{out} = \frac{U_{in}}{R_1 + R_2}R_2
$$

  

$$
U_{out} = \frac{R_2}{R_1 + R_2}U_{in}
$$

  

This shows that the output is a fixed fraction of the input voltage, determined by the ratio of the two resistors.
