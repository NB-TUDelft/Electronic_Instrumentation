# Resistance recap

## Core laws and concepts


In electronics, resistance is the opposition to the flow of electric current in an electrical circuit. Resistance is measured in ohms ($\Omega$). There is also a quantity that measures how easily current can flow through a material, called conductance $G$, with units of siemens ($\mathrm{S}$). In basic electronics and electrotechnics, there are 3 main laws that are essential for analyzing, describing, and calculating parameters of electronic circuits. These are Ohm's law and Kirchhoff's laws, 1st (junction rule) and 2nd (loop rule). It is important to keep the idea of these laws in mind as a concept, not a bunch of formulas. Someone smart used to say these are enough to get a PhD in electrical engineering.

Ohm's law states that the voltage $U$ across a resistor is proportional to the current $I$ through it: $U = RI$. This means that for a fixed resistance $R$, a larger voltage $U$ causes a larger current $I$. Kirchhoff's current law (junction rule) states that the sum of currents entering a node must equal the sum of currents leaving it. Kirchhoff's voltage law (loop rule) states that the sum of all voltage rises and drops around a closed loop must be zero. These laws express conservation of charge and conservation of energy in electrical circuits.

All electronic circuits and further formulas can be derived from these laws. Let us first look at the parallel case, then the series case, and finally use that result to derive the voltage divider formula.


## Parallel resistors

```{figure} ../images/paralel.svg
Paralel resistors
```


For two resistors in parallel, the voltage across both resistors is the same:

$$ U = U_1 = U_2 $$

Using Ohm's law, the current in each branch is:

$$ I_1 = \frac{U}{R_1}, \qquad I_2 = \frac{U}{R_2} $$

Kirchhoff's current law tells us that the total current is the sum of the branch currents:

$$ I = I_1 + I_2 $$

Substituting the expressions for $I_1$ and $I_2$ gives:

$$ I = \frac{U}{R_1} + \frac{U}{R_2} $$

$$ I = U\left(\frac{1}{R_1} + \frac{1}{R_2}\right) $$

For the equivalent resistance $R_{eq}$ of the same network, we also have the following:

$$ I = \frac{U}{R_{eq}} $$

So:

$$ \frac{U}{R_{eq}} = U\left(\frac{1}{R_1} + \frac{1}{R_2}\right) $$

Dividing both sides by $U$ gives:

$$ \frac{1}{R_{eq}} = \frac{1}{R_1} + \frac{1}{R_2} $$

For two resistors, this can be rewritten as:

$$ R_{eq} = \frac{R_1R_2}{R_1 + R_2} $$



## Series resistors

```{figure} ../images/series.svg
---
height: 400px
name: series
---
Series resistors
```

Now compare this with the series case. For two resistors in series, the same current flows through both resistors:

$$ I = I_1 = I_2 $$

Kirchhoff's voltage law tells us that the total voltage is the sum of the voltage drops:

$$ U = U_1 + U_2 $$

Using Ohm's law for each resistor gives:

$$ U = IR_1 + IR_2 $$

$$ U = I(R_1 + R_2) $$

For the equivalent resistance $R_{eq}$ we also write:

$$ U = IR_{eq} $$

Therefore:

$$ R_{eq} = R_1 + R_2 $$


## Voltage divider


This series result leads smoothly to the voltage divider formula. If $R_1$ and $R_2$ are connected in series across an input voltage $U_{in}$ and the output is measured across $R_2$, then the same current flows through both resistors:

$$ I = \frac{U_{in}}{R_1 + R_2} $$

The output voltage is the voltage across $R_2$:

$$ U_{out} = IR_2 $$

Substituting the expression for $I$ gives the following:

$$ U_{out} = \frac{U_{in}}{R_1 + R_2}R_2 $$

$$ U_{out} = \frac{R_2}{R_1 + R_2}U_{in} $$

This shows that the output is a fixed fraction of the input voltage, determined by the ratio of the two resistors.

## Resistance measurements

There are many ways to measure resistance; here are some of the most common ones:
- direct resistance measurement (using an ohmmeter), often implemented as a 2-wire resistance measurement
- indirect resistance measurement (Voltmeter-Ammeter Method)
- bridge resistance measurement (eg.  Wheatstone Bridge method)
- comparative resistance measurement



### Direct resistance measurement

Direct resistance measurement is performed with an ohmmeter or a digital multimeter (DMM) set to resistance mode. The method is based on Ohm's law: the instrument applies a DC excitation and determines the unknown resistance from the resulting current or voltage. In analog ohmmeters, resistance is converted into a current reading. In digital ohmmeters, resistance is typically determined from a measured voltage. In both cases, the measurement circuit uses a constant excitation, so the measured quantity depends on the unknown resistance $R_X$. Analog ohmmeters are simple and not oftenly seen these days, digital ohmmeters usually offer better accuracy, and they are oftenly a part of a DMM.

#### 2-wire method

The most common direct method in basic laboratory work is the 2-wire method. The same two leads are used both to force the test current and to measure the resulting voltage across the resistor. A modern DMM then calculates the resistance from Ohm's law.

For an ideal direct measurement,

$$ R = \frac{V}{I} $$

In a real 2-wire measurement, the meter does not see only the unknown resistor. The resistance of the test leads is also part of the measured path, so the measured value becomes approximately

$$ R_{meas} = R_X + R_{lead_1} + R_{lead_2} $$

In practice, contact resistance can also add extra error. This is why the 2-wire method is simple and convenient, but it becomes less accurate for very small resistances. For low resistances, the resistance of the leads may be comparable to the resistor you want to measure.

#### Practical notes

- Turn off power before measuring resistance.
- If possible, isolate the component from the circuit before measuring it.
- If your meter has a relative or null function, short the probes first and subtract the lead resistance.
- For low-resistance or high-accuracy measurements, a [4-wire (Kelvin) method](https://en.wikipedia.org/wiki/Four-terminal_sensing) is usually better.


### Indirect resistance measurement

The Voltmeter-Ammeter method, also called the technical method, is used primarily for measuring nonlinear resistances, but it can also be used for linear resistances. The method uses a voltmeter and an ammeter, and the resistance is calculated from Ohm's law:

$$ R_X = \frac{U}{I} $$

In practice, real instruments are not ideal. The ammeter has internal resistance $R_A$ and the voltmeter has finite input resistance $R_V$. Because of this, it is not possible to measure both $I_X$ and $U_X$ perfectly at the same time in one simple connection. Therefore, two circuit arrangements are used.

In the formulas below, $R_{Xp}$ means the measured (calculated) resistance value obtained from the meter readings, while $R_X$ is the true resistance of the tested resistor.

#### Circuit 1: Current-correct

```{figure} ../images/current.svg
---
name: Current
height: 400px
---
Properly measured current
```

In this arrangement, the ammeter reading is correct for the resistor current, so $I_A = I_X$. However, the voltmeter reads the resistor voltage plus the voltage drop on the ammeter internal resistance $R_A$.

**Result: the measured/calculated resistance is too big**.

$$ R_{Xp} = \frac{U_V}{I_A} = R_X + R_A $$

Absolute method error:

$$ \Delta_i = R_{Xp} - R_X = R_A > 0 $$

Relative method error:

$$ \delta_i = \frac{\Delta_i}{R_X} = \frac{R_A}{R_X} $$

This arrangement should be used for large resistances, where

$$ R_X \gg R_A $$

#### Circuit 2: Voltage-correct

```{figure} ../images/voltage.svg
---
name: Voltage
height: 400px
---
Properly measured voltage
```

In this arrangement, the voltmeter is connected directly across $R_X$, so the voltage reading is correct for the resistor. The ammeter, however, measures total current

$$ I_A = I_X + I_V $$

because part of the current flows through the voltmeter input resistance $R_V$.

**Result: the measured/calculated resistance is too small**.

$$ R_{Xp} = \frac{U_V}{I_A} = \frac{R_X R_V}{R_X + R_V} $$

Absolute method error:

$$ \Delta_v = R_{Xp} - R_X = -\frac{R_X^2}{R_X + R_V} < 0 $$

Relative method error:

$$ \delta_v = \frac{\Delta_v}{R_X} = -\frac{R_X}{R_X + R_V} $$

This arrangement should be used for small resistances, where

$$ R_X \ll R_V $$

#### Choosing the better arrangement

A practical boundary value is the limiting resistance

$$ R_{gr} = \sqrt{R_A R_V} $$

If the expected $R_X$ is below $R_{gr}$, use the correctly measured voltage arrangement. If it is above $R_{gr}$, use the correctly measured current arrangement.

In both arrangements, method errors are systematic (they have a defined sign), so they can be corrected when reporting the final result.
